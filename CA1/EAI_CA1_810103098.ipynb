# <span style="color:#2c3e50">Part 1: Logarithmic Number System (LNS) Conversion & Validation</span>

## <span style="color:#2980b9">1.1 Introduction</span>
In this section, we implement the core functions for the **Logarithmic Number System (LNS)**. LNS is an alternative to fixed-point and floating-point representations, where a number $X$ is represented by its sign and the logarithm of its absolute value.

The representation is defined as:
$$ \bar{X} = (z_x, s_x, m_x) $$

Where:
*   $z_x$: **Zero flag** (True if $X=0$). Since $\log(0) = -\infty$, zero requires special handling.
*   $s_x$: **Sign bit** (0 for positive, 1 for negative).
*   $m_x$: **Fixed-point mantissa**, representing $\log_{base}(|X|)$ in two's complement format.

The value of the number is given by:
$$ X = (-1)^{s_x} \cdot (1-z_x) \cdot \text{base}^{(m_x \cdot 2^{-f})} $$

## <span style="color:#2980b9">1.2 Implementation Details</span>
We define a class `LNSNumber` and three key functions:
1.  **`real_to_lns(x, ...)`**: Converts a real floating-point number to LNS.
2.  **`lns_to_real(lns, ...)`**: Converts an LNS number back to the real domain.
3.  **`get_lns_range(...)`**: Calculates the minimum and maximum positive representable values.

### <span style="color:#e67e22">Parameters for Validation</span>
To validate our implementation, we use the following parameters as specified in the problem statement:
*   **Base:** $b = 2$
*   **Total Bits:** $W = 6$
*   **Fractional Bits:** $f = 3$



### LNS (Logarithmic Number System) Conversion Functions

```python
import numpy as np
from typing import Tuple, List
import warnings
warnings.filterwarnings('ignore')

class LNSNumber:
    """
    Represents a number in Logarithmic Number System (LNS).

    Format: (z, s, m) where:
        - z (bool): Zero flag (True if number is zero)
        - s (bool): Sign flag (True if negative)
        - m (int): Fixed-point representation of log_base(|x|) in two's complement
    """

    def __init__(self, z: bool, s: bool, m: int, base: float, W: int, f: int):
        """
        Initialize an LNS number.

        Args:
            z: Zero flag
            s: Sign flag (True for negative)
            m: Fixed-point mantissa (log value in fixed-point)
            base: Logarithm base
            W: Total bit width for mantissa
            f: Number of fractional bits
        """
        self.z = z
        self.s = s
        self.m = m
        self.base = base
        self.W = W
        self.f = f

    def __repr__(self):
        if self.z:
            return f"LNS(zero=True, base={self.base}, W={self.W}, f={self.f})"
        sign_str = "-" if self.s else "+"
        # Convert fixed-point m to actual log value
        log_val = self.m / (2 ** self.f)
        return f"LNS(z={self.z}, s='{sign_str}', m={self.m}, log_val={log_val:.4f}, base={self.base}, W={self.W}, f={self.f})"


def get_lns_range(base: float, W: int, f: int) -> Tuple[float, float]:
    """
    Calculate the range of positive real numbers representable in LNS.

    In LNS with W-bit two's complement mantissa and f fractional bits:
    - Mantissa range: [-2^(W-1), 2^(W-1) - 1] as integers
    - Actual log values: [m_min / 2^f, m_max / 2^f]
    - Real number range: [base^(m_min/2^f), base^(m_max/2^f)]

    Args:
        base: Logarithm base
        W: Total bit width for mantissa
        f: Number of fractional bits

    Returns:
        Tuple of (min_positive, max_positive) representable values
    """
    # Two's complement range for W bits
    m_min_int = -(2 ** (W - 1))       # Minimum integer value
    m_max_int = (2 ** (W - 1)) - 1    # Maximum integer value

    # Convert to actual log values (divide by 2^f for fixed-point)
    m_min_real = m_min_int / (2 ** f)
    m_max_real = m_max_int / (2 ** f)

    # Calculate real number range
    min_positive = base ** m_min_real
    max_positive = base ** m_max_real

    return (min_positive, max_positive)


def real_to_lns(x: float, base: float, W: int, f: int) -> LNSNumber:
    """
    Convert a real number to LNS representation.

    Args:
        x: Real number to convert
        base: Logarithm base
        W: Total bit width for mantissa
        f: Number of fractional bits

    Returns:
        LNSNumber object representing x in LNS format
    """
    # Handle zero case
    if x == 0:
        return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)

    # Determine sign
    s = x < 0
    abs_x = abs(x)

    # Calculate logarithm: log_base(|x|)
    log_val = np.log(abs_x) / np.log(base)

    # Convert to fixed-point representation
    # Multiply by 2^f and round to nearest integer
    m_float = log_val * (2 ** f)
    m_int = int(np.round(m_float))

    # Clamp to valid range for W-bit two's complement
    m_min = -(2 ** (W - 1))
    m_max = (2 ** (W - 1)) - 1
    m_int = max(m_min, min(m_max, m_int))

    return LNSNumber(z=False, s=s, m=m_int, base=base, W=W, f=f)


def lns_to_real(lns_num: LNSNumber) -> float:
    """
    Convert an LNS number back to real number representation.

    Args:
        lns_num: LNSNumber object

    Returns:
        Real number value
    """
    # Handle zero case
    if lns_num.z:
        return 0.0

    # Convert fixed-point mantissa to actual log value
    log_val = lns_num.m / (2 ** lns_num.f)

    # Calculate absolute value: base^log_val
    abs_val = lns_num.base ** log_val

    # Apply sign
    if lns_num.s:
        return -abs_val
    return abs_val


def calculate_errors(original: float, converted: float) -> Tuple[float, float]:
    """
    Calculate absolute and relative errors.

    Args:
        original: Original real number
        converted: Converted value after LNS round-trip

    Returns:
        Tuple of (absolute_error, relative_error)
    """
    abs_error = abs(converted - original)

    # Handle relative error for zero
    if original == 0:
        rel_error = 0.0 if converted == 0 else float('inf')
    else:
        rel_error = abs_error / abs(original)

    return (abs_error, rel_error)
```

    *   **0.0** is handled perfectly via the $z$ flag.
    *   Negative numbers (**-2.0**, **-7.5**) are handled correctly using the $s$ flag, with errors matching their positive counterparts.

## <span style="color: #ffa502;">Part 2: Conversion Error Analysis with Different Parameters</span>

### **2.1 Problem Statement**
The precision and dynamic range of an LNS system depend heavily on the **Base**, **Total Word Length ($W$)**, and **Fractional Bits ($f$)**. In this section, we analyze the impact of these parameters by comparing four distinct configurations:

1.  $\text{Base} = \sqrt{2}, W=8, f=4$
2.  $\text{Base} = 2, W=7, f=4$
3.  $\text{Base} = 4, W=6, f=4$
4.  $\text{Base} = 16, W=5, f=4$

### **2.2 Methodology**
For each configuration, we perform the following steps:
1.  **Calculate Range:** Determine the Minimum ($X_{min}$) and Maximum ($X_{max}$) representable positive numbers.
2.  **Sampling:** Generate $1000$ uniformly distributed numbers within $[X_{min}, X_{max}]$.
3.  **Error Computation:** Convert samples to LNS and back to Real, then compute:
    *   **Absolute Error:** $|X_{real} - X_{converted}|$
    *   **Relative Error:** $\frac{|X_{real} - X_{converted}|}{|X_{real}|}$
4.  **Trade-off Analysis:** Visualize how reducing $W$ while increasing the Base affects the **Dynamic Range** vs. **Precision**.

</div>

```python
# This section analyzes the error introduced by converting real numbers to LNS
# representation with different parameter sets (base, W, f).
# We will:
# 1. Calculate the representable range for each parameter set
# 2. Generate uniformly distributed test numbers in each range
# 3. Compute and plot absolute and relative errors
# 4. Compare and analyze the trade-offs between different configurations
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt
from typing import Tuple, List, Dict

# The LNS conversion functions and LNSNumber class are defined in cell 9QkOhczB3XYh.
# We will use those definitions directly here.

# -----------------------------------------------------------------------------
# Error Analysis Functions
# -----------------------------------------------------------------------------

def compute_conversion_errors(test_numbers: np.ndarray, base: float, W: int, f: int) -> Tuple[np.ndarray, np.ndarray]:
    """
    Compute absolute and relative errors for converting numbers to LNS and back.

    Args:
        test_numbers: Array of positive real numbers to test
        base: Logarithm base
        W: Total bit width
        f: Number of fractional bits

    Returns:
        Tuple of (absolute_errors, relative_errors)
    """
    absolute_errors = []
    relative_errors = []

    for x in test_numbers:
        # Convert to LNS
        lns_num = real_to_lns(x, base, W, f)
        # Convert back to real
        x_reconstructed = lns_to_real(lns_num)

        # Calculate errors using the globally defined function
        abs_error, rel_error = calculate_errors(x, x_reconstructed)

        absolute_errors.append(abs_error)
        relative_errors.append(rel_error)

    return np.array(absolute_errors), np.array(relative_errors)


def analyze_parameter_set(base: float, W: int, f: int, num_samples: int = 1000) -> Dict:
    """
    Perform complete error analysis for a given parameter set.

    Args:
        base: Logarithm base
        W: Total bit width
        f: Number of fractional bits
        num_samples: Number of test samples to generate

    Returns:
        Dictionary containing analysis results
    """
    # Get representable range using the globally defined function
    min_val, max_val = get_lns_range(base, W, f)

    # Generate uniformly distributed test numbers
    test_numbers = np.linspace(min_val, max_val, num_samples)

    # Compute errors
    abs_errors, rel_errors = compute_conversion_errors(test_numbers, base, W, f)

    # Calculate statistics
    results = {
        'base': base,
        'W': W,
        'f': f,
        'min_val': min_val,
        'max_val': max_val,
        'range_ratio': max_val / min_val,
        'test_numbers': test_numbers,
        'absolute_errors': abs_errors,
        'relative_errors': rel_errors,
        'mean_abs_error': np.mean(abs_errors),
        'mean_rel_error': np.mean(rel_errors),
        'max_abs_error': np.max(abs_errors),
        'max_rel_error': np.max(rel_errors),
        'std_rel_error': np.std(rel_errors)
    }

    return results


# Main Analysis

# Define parameter sets as specified in the problem
parameter_sets = [
    (np.sqrt(2), 8, 4),  # base=√2, W=8, f=4
    (2, 7, 4),           # base=2,  W=7, f=4
    (4, 6, 4),           # base=4,  W=6, f=4
    (16, 5, 4)           # base=16, W=5, f=4
]

# Analyze each parameter set
results_list = []
for base, W, f in parameter_sets:
    print(f"\n{'='*70}")
    print(f"Analyzing: base={base:.4f}, W={W}, f={f}")
    print('='*70)

    results = analyze_parameter_set(base, W, f, num_samples=1000)
    results_list.append(results)

    print(f"Representable Range: [{results['min_val']:.6e}, {results['max_val']:.6e}]")
    print(f"Dynamic Range Ratio: {results['range_ratio']:.2e}")
    print(f"Mean Absolute Error: {results['mean_abs_error']:.6e}")
    print(f"Mean Relative Error: {results['mean_rel_error']:.6f} ({results['mean_rel_error']*100:.4f}%)")
    print(f"Max Relative Error:  {results['max_rel_error']:.6f} ({results['max_rel_error']*100:.4f}%)")
```

# **Part 3: LNS Addition Operations with Different Approximations**

# <span style="color:#8e44ad">Part 3: LNS Addition Operations</span>

## <span style="color:#9b59b6">3.1 The Addition Challenge</span>
Unlike multiplication, which is trivial in LNS ($ \log(A \cdot B) = \log A + \log B $), addition is non-linear and computationally expensive.

Given two numbers $X$ and $Y$ (where $X > Y$), their sum in LNS is:
$$ \log_b(X \pm Y) = \log_b(X) + \log_b(1 \pm \frac{Y}{X}) $$

Let $m_1 = \log_b X$ and $m_2 = \log_b Y$. Defining $d = |m_1 - m_2|$, the addition formula becomes:
$$ \text{Sum}_{LNS} = \max(m_1, m_2) + \Delta_{\pm}(d) $$

Where the correction term $\Delta$ is:
$$ \Delta_{\pm}(d) = \log_b(1 \pm b^{-d}) $$

## <span style="color:#9b59b6">3.2 Implemented Methods</span>
We implement three distinct methods to handle the $\Delta$ term:

1.  **Exact Computation:** Calculates $\log_b(1 \pm b^{-d})$ precisely using standard math libraries. This serves as our "Ground Truth".
2.  **LUT Approximation:** Uses a pre-computed **Look-Up Table (LUT)** with **32 entries** to find the nearest stored $\Delta$ value for a given $d$. This simulates hardware where memory is limited.
3.  **Bit-Shift Approximation:** Uses linear approximations (Taylor series) to estimate $\log(1+x) \approx x$. In hardware, this corresponds to simple bit-shifting operations.

### <span style="color:#e67e22">Test Parameters</span>
*   **Base:** $b=2$
*   **Word Length:** $W=8$ ($f=4$)
*   **LUT Size:** 32 entries

### LNS ADDITION OPERATIONS

```python
# This section implements three addition methods in LNS domain:
# 1. Exact computation of Δ term
# 2. LUT-based approximation
# 3. Bit-shift based approximation
#
# The key challenge in LNS addition is computing:
# Δ± = log_base(1 ± 2^(-|m1 - m2|))
# =============================================================================

import numpy as np
from typing import Tuple, Dict, List

# -----------------------------------------------------------------------------
# Core LNS Functions (from previous parts)
# -----------------------------------------------------------------------------

class LNSNumber:
    """
    Represents a number in Logarithmic Number System (LNS).
    Format: (z, s, m) where:
        - z: Zero flag (True if number is zero)
        - s: Sign flag (True if negative)
        - m: Fixed-point log value
    """
    def __init__(self, z: bool, s: bool, m: int, base: float, W: int, f: int):
        self.z = z
        self.s = s
        self.m = m
        self.base = base
        self.W = W
        self.f = f

    def get_log_value(self) -> float:
        """Convert fixed-point m to actual log value."""
        if self.z:
            return float('-inf')
        return self.m / (2 ** self.f)

    def __repr__(self):
        if self.z:
            return "LNS(0)"
        sign = "-" if self.s else "+"
        return f"LNS({sign}, m={self.m}, log={self.get_log_value():.4f})"


def get_lns_range(base: float, W: int, f: int) -> Tuple[float, float]:
    """Calculate the representable range for positive numbers."""
    m_min = -(2 ** (W - 1))
    m_max = (2 ** (W - 1)) - 1
    min_val = base ** (m_min / (2 ** f))
    max_val = base ** (m_max / (2 ** f))
    return (min_val, max_val)


def real_to_lns(x: float, base: float, W: int, f: int) -> LNSNumber:
    """Convert real number to LNS representation."""
    if x == 0:
        return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)

    s = x < 0
    log_val = np.log(abs(x)) / np.log(base)
    m_float = log_val * (2 ** f)
    m_int = int(np.round(m_float))

    # Clamp to valid range
    m_min = -(2 ** (W - 1))
    m_max = (2 ** (W - 1)) - 1
    m_int = max(m_min, min(m_max, m_int))

    return LNSNumber(z=False, s=s, m=m_int, base=base, W=W, f=f)


def lns_to_real(lns_num: LNSNumber) -> float:
    """Convert LNS representation back to real number."""
    if lns_num.z:
        return 0.0

    log_val = lns_num.m / (2 ** lns_num.f)
    abs_val = lns_num.base ** log_val
    return -abs_val if lns_num.s else abs_val
```

### LUT Generation for Δ+ and Δ- terms

```python
def generate_delta_lut(base: float, W: int, f: int, lut_size: int = 32) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Pre-compute LUT for Delta+ and Delta- approximations.

    Delta+ = log_base(1 + base^(-d))  for addition (same signs)
    Delta- = log_base(1 - base^(-d))  for subtraction (different signs)

    Note: d represents |m1 - m2| in the LOG DOMAIN (not fixed-point)
    The LUT covers the practical range of d values.

    Args:
        base: Logarithm base
        W: Total bit width
        f: Fractional bits
        lut_size: Number of entries in each LUT

    Returns:
        Tuple of (d_values, delta_plus, delta_minus) arrays
    """
    # d is the difference in log domain: |log(|a|) - log(|b|)|
    # Maximum practical d value based on representable range
    max_d = (2 ** W) / (2 ** f)  # Maximum difference in log domain

    # Generate d values with finer resolution near 0 (where delta changes rapidly)
    d_values = np.linspace(0, max_d, lut_size)

    # Compute Delta+ = log_base(1 + base^(-d))
    delta_plus = np.log(1 + np.power(base, -d_values)) / np.log(base)

    # Compute Delta- = log_base(1 - base^(-d))
    # Note: For d=0, this is -inf (log(0)), handle specially
    delta_minus = np.zeros(lut_size)
    for i, d in enumerate(d_values):
        if d == 0:
            delta_minus[i] = float('-inf')
        else:
            val = 1 - np.power(base, -d)
            if val > 0:
                delta_minus[i] = np.log(val) / np.log(base)
            else:
                delta_minus[i] = float('-inf')

    return d_values, delta_plus, delta_minus


def lookup_delta(d: float, d_values: np.ndarray, delta_lut: np.ndarray) -> float:
    """
    Look up delta value from pre-computed LUT using nearest neighbor interpolation.

    Args:
        d: The difference |m1 - m2| in log domain
        d_values: Array of d values in LUT
        delta_lut: Array of corresponding delta values

    Returns:
        Approximated delta value
    """
    # Clamp d to LUT range
    d_clamped = np.clip(d, d_values[0], d_values[-1])
    # Find nearest index
    idx = np.argmin(np.abs(d_values - d_clamped))
    return delta_lut[idx]
```

### Three Addition Methods

```python
def lns_add_exact(a: LNSNumber, b: LNSNumber) -> LNSNumber:
    """
    LNS addition with exact Delta computation.

    For same signs:   result_log = max(m1,m2) + log_base(1 + base^(-|m1-m2|))
    For diff signs:   result_log = max(m1,m2) + log_base(1 - base^(-|m1-m2|))

    Args:
        a, b: LNS numbers to add

    Returns:
        LNS representation of a + b
    """
    base, W, f = a.base, a.W, a.f

    # Handle zero cases
    if a.z and b.z:
        return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)
    if a.z:
        return LNSNumber(z=b.z, s=b.s, m=b.m, base=base, W=W, f=f)
    if b.z:
        return LNSNumber(z=a.z, s=a.s, m=a.m, base=base, W=W, f=f)

    # Get log values (in real log domain, not fixed-point)
    m_a = a.get_log_value()
    m_b = b.get_log_value()

    # Determine which has larger magnitude
    if m_a >= m_b:
        m_max, m_min = m_a, m_b
        s_max = a.s
    else:
        m_max, m_min = m_b, m_a
        s_max = b.s

    # Calculate difference in log domain
    d = m_max - m_min

    # Same signs: addition -> use Delta+
    if a.s == b.s:
        # Delta+ = log_base(1 + base^(-d))
        delta = np.log(1 + base ** (-d)) / np.log(base)
        result_log = m_max + delta
        result_sign = a.s
    else:
        # Different signs: subtraction -> use Delta-
        if d == 0:
            # Equal magnitudes, opposite signs -> zero
            return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)

        val = 1 - base ** (-d)
        if val <= 0:
            return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)

        delta = np.log(val) / np.log(base)
        result_log = m_max + delta
        result_sign = s_max

    # Convert back to fixed-point
    m_result = int(np.round(result_log * (2 ** f)))

    # Clamp to valid range
    m_min_val = -(2 ** (W - 1))
    m_max_val = (2 ** (W - 1)) - 1
    m_result = max(m_min_val, min(m_max_val, m_result))

    return LNSNumber(z=False, s=result_sign, m=m_result, base=base, W=W, f=f)


def lns_add_lut(a: LNSNumber, b: LNSNumber,
                d_values: np.ndarray, delta_plus: np.ndarray, delta_minus: np.ndarray) -> LNSNumber:
    """
    LNS addition using LUT-based approximation for Delta.

    Args:
        a, b: LNS numbers to add
        d_values, delta_plus, delta_minus: Pre-computed LUT arrays

    Returns:
        LNS representation of a + b (approximated)
    """
    base, W, f = a.base, a.W, a.f

    # Handle zero cases
    if a.z and b.z:
        return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)
    if a.z:
        return LNSNumber(z=b.z, s=b.s, m=b.m, base=base, W=W, f=f)
    if b.z:
        return LNSNumber(z=a.z, s=a.s, m=a.m, base=base, W=W, f=f)

    # Get log values
    m_a = a.get_log_value()
    m_b = b.get_log_value()

    # Determine which is larger
    if m_a >= m_b:
        m_max, m_min = m_a, m_b
        s_max = a.s
    else:
        m_max, m_min = m_b, m_a
        s_max = b.s

    d = m_max - m_min

    # Same signs: use Delta+
    if a.s == b.s:
        delta = lookup_delta(d, d_values, delta_plus)
        result_log = m_max + delta
        result_sign = a.s
    else:
        # Different signs: use Delta-
        delta = lookup_delta(d, d_values, delta_minus)
        if np.isinf(delta) and delta < 0:
            return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)
        result_log = m_max + delta
        result_sign = s_max

    # Convert to fixed-point
    m_result = int(np.round(result_log * (2 ** f)))
    m_min_val = -(2 ** (W - 1))
    m_max_val = (2 ** (W - 1)) - 1
    m_result = max(m_min_val, min(m_max_val, m_result))

    return LNSNumber(z=False, s=result_sign, m=m_result, base=base, W=W, f=f)


def lns_add_shift(a: LNSNumber, b: LNSNumber) -> LNSNumber:
    """
    LNS addition using bit-shift approximation.

    Approximations used:
    - For large d: log_base(1 + base^(-d)) ≈ base^(-d) / ln(base)
    - For small d: log_base(1 + base^(-d)) ≈ log_base(2) when d ≈ 0

    The key idea is that for large d, the correction term becomes negligible
    and can be approximated with simple bit shifts.

    Args:
        a, b: LNS numbers to add

    Returns:
        LNS representation of a + b (approximated)
    """
    base, W, f = a.base, a.W, a.f

    # Handle zero cases
    if a.z and b.z:
        return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)
    if a.z:
        return LNSNumber(z=b.z, s=b.s, m=b.m, base=base, W=W, f=f)
    if b.z:
        return LNSNumber(z=a.z, s=a.s, m=a.m, base=base, W=W, f=f)

    # Get log values
    m_a = a.get_log_value()
    m_b = b.get_log_value()

    # Determine which is larger
    if m_a >= m_b:
        m_max, m_min = m_a, m_b
        s_max = a.s
    else:
        m_max, m_min = m_b, m_a
        s_max = b.s

    d = m_max - m_min

    # Bit-shift approximation thresholds
    # For base=2: when d > 1, the correction term base^(-d) < 0.5
    threshold_small = 0.5
    threshold_large = 4.0  # Beyond this, correction is negligible

    if a.s == b.s:
        # Same signs: addition
        if d < threshold_small:
            # Small d: Delta+ ≈ log_base(2) = 1/log_2(base)
            delta = 1.0 / np.log2(base)
        elif d < threshold_large:
            # Medium d: Delta+ ≈ base^(-d) / ln(base)
            delta = (base ** (-d)) / np.log(base)
        else:
            # Large d: Delta+ ≈ 0 (negligible correction)
            delta = 0.0
        result_log = m_max + delta
        result_sign = a.s
    else:
        # Different signs: subtraction
        if d < 0.1:
            # Very close values -> result approaches zero
            return LNSNumber(z=True, s=False, m=0, base=base, W=W, f=f)
        elif d < threshold_large:
            # Delta- ≈ -base^(-d) / ln(base) for moderate d
            delta = -(base ** (-d)) / np.log(base)
        else:
            # Large d: Delta- ≈ 0
            delta = 0.0
        result_log = m_max + delta
        result_sign = s_max

    # Convert to fixed-point
    m_result = int(np.round(result_log * (2 ** f)))
    m_min_val = -(2 ** (W - 1))
    m_max_val = (2 ** (W - 1)) - 1
    m_result = max(m_min_val, min(m_max_val, m_result))

    return LNSNumber(z=False, s=result_sign, m=m_result, base=base, W=W, f=f)
```

### STATISTICAL ANALYSIS OF LNS ADDITION ERROR

```python
# In this section, we:
# 1. Generate 1000 random positive number pairs within representable range
# 2. Compute addition using all three methods (Exact, LUT, Shift)
# 3. Analyze errors as a function of |m_a - m_b|
# 4. Study the effect of LUT size on approximation error
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt
from typing import Tuple, Dict, List

# -----------------------------------------------------------------------------
# LNS Core Functions (from previous sections)
# -----------------------------------------------------------------------------

def get_signed_mantissa(m: int, W: int) -> int:
    """Convert two's complement mantissa to signed integer."""
    if m >= (1 << (W - 1)):
        return m - (1 << W)
    return m


def to_twos_complement(val: int, W: int) -> int:
    """Convert signed integer to two's complement representation."""
    if val < 0:
        return (1 << W) + val
    return val


def real_to_lns(x: float, base: float, W: int, f: int) -> Tuple[int, int, int]:
    """Convert real number to LNS representation (z, s, m)."""
    if x == 0:
        return (1, 0, 0)

    s = 0 if x > 0 else 1
    log_val = np.log(abs(x)) / np.log(base)

    scale = 2 ** f
    m_int = int(np.round(log_val * scale))

    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_int = max(min_val, min(max_val, m_int))

    if m_int < 0:
        m = (1 << W) + m_int
    else:
        m = m_int

    return (0, s, m)


def lns_to_real(z: int, s: int, m: int, base: float, W: int, f: int) -> float:
    """Convert LNS representation back to real number."""
    if z == 1:
        return 0.0

    m_signed = get_signed_mantissa(m, W)
    log_val = m_signed / (2 ** f)
    abs_val = base ** log_val

    return -abs_val if s == 1 else abs_val


def get_lns_range(base: float, W: int, f: int) -> Tuple[float, float]:
    """Calculate representable range for positive numbers."""
    min_log = -(2 ** (W - 1)) / (2 ** f)
    max_log = ((2 ** (W - 1)) - 1) / (2 ** f)
    return (base ** min_log, base ** max_log)
```

### LUT Class for Delta Approximation

```python

class LNS_LUT:
    """Lookup table for LNS addition delta values."""

    def __init__(self, base: float, W: int, f: int, lut_size: int = 32):
        self.base = base
        self.W = W
        self.f = f
        self.lut_size = lut_size

        # Maximum d value to cover in LUT
        self.d_max = W - f + 4
        self.d_step = self.d_max / (lut_size - 1)

        # Generate d values for LUT indexing
        self.d_values = np.linspace(0, self.d_max, lut_size)

        # Compute Δ+ = log_base(1 + 2^(-d))
        self.delta_plus = np.log(1 + 2 ** (-self.d_values)) / np.log(base)

        # Compute Δ- = log_base(1 - 2^(-d))
        with np.errstate(divide='ignore', invalid='ignore'):
            self.delta_minus = np.log(1 - 2 ** (-self.d_values)) / np.log(base)
        self.delta_minus[0] = -np.inf  # log(0) = -inf

    def lookup_delta_plus(self, d: float) -> float:
        """Look up Δ+ for given d value."""
        d = max(0, min(d, self.d_max))
        idx = int(round(d / self.d_step))
        idx = max(0, min(idx, self.lut_size - 1))
        return self.delta_plus[idx]

    def lookup_delta_minus(self, d: float) -> float:
        """Look up Δ- for given d value."""
        d = max(0, min(d, self.d_max))
        idx = int(round(d / self.d_step))
        idx = max(0, min(idx, self.lut_size - 1))
        return self.delta_minus[idx]
```

### LNS Addition Functions

```python
def lns_add_exact(lns1: Tuple, lns2: Tuple, base: float, W: int, f: int) -> Tuple[int, int, int]:
    """LNS addition with exact delta computation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    # Convert to real log values
    m1_real = get_signed_mantissa(m1, W) / (2 ** f)
    m2_real = get_signed_mantissa(m2, W) / (2 ** f)

    # Find max and compute d
    if m1_real >= m2_real:
        m_max, s_max, s_min = m1_real, s1, s2
    else:
        m_max, s_max, s_min = m2_real, s2, s1

    d = abs(m1_real - m2_real)

    # Compute delta based on signs
    if s_max == s_min:
        delta = np.log(1 + base ** (-d)) / np.log(base)
        result_sign = s_max
    else:
        inner = 1 - base ** (-d)
        if inner <= 0:
            return (1, 0, 0)
        delta = np.log(inner) / np.log(base)
        result_sign = s_max

    # Compute result mantissa
    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    # Clamp and convert
    min_val, max_val = -(2 ** (W - 1)), (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))


def lns_add_lut(lns1: Tuple, lns2: Tuple, base: float, W: int, f: int, lut: LNS_LUT) -> Tuple[int, int, int]:
    """LNS addition using LUT-based delta approximation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    m1_real = get_signed_mantissa(m1, W) / (2 ** f)
    m2_real = get_signed_mantissa(m2, W) / (2 ** f)

    if m1_real >= m2_real:
        m_max, s_max, s_min = m1_real, s1, s2
    else:
        m_max, s_max, s_min = m2_real, s2, s1

    d = abs(m1_real - m2_real)

    if s_max == s_min:
        delta = lut.lookup_delta_plus(d)
        result_sign = s_max
    else:
        delta = lut.lookup_delta_minus(d)
        if np.isinf(delta) and delta < 0:
            return (1, 0, 0)
        result_sign = s_max

    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    min_val, max_val = -(2 ** (W - 1)), (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))


def lns_add_shift(lns1: Tuple, lns2: Tuple, base: float, W: int, f: int) -> Tuple[int, int, int]:
    """LNS addition using bit-shift approximation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    m1_real = get_signed_mantissa(m1, W) / (2 ** f)
    m2_real = get_signed_mantissa(m2, W) / (2 ** f)

    if m1_real >= m2_real:
        m_max, s_max, s_min = m1_real, s1, s2
    else:
        m_max, s_max, s_min = m2_real, s2, s1

    d = abs(m1_real - m2_real)
    d_int = int(np.floor(d))

    if s_max == s_min:
        # Approximation: Δ+ ≈ 2^(-d) for large d
        if d_int >= W:
            delta = 0
        else:
            # Use first-order approximation
            delta = np.log(1 + 2**(-d_int)) / np.log(base) if d_int >= 0 else 1/np.log(base)
        result_sign = s_max
    else:
        if d < 0.1:
            return (1, 0, 0)
        if d_int >= W:
            delta = 0
        else:
            inner = 1 - 2**(-d_int) if d_int > 0 else 0
            if inner <= 0:
                return (1, 0, 0)
            delta = np.log(inner) / np.log(base)
        result_sign = s_max

    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    min_val, max_val = -(2 ** (W - 1)), (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))
```

# <span style="color:#2c3e50">Part 5: Impact of Base Selection on LNS Performance</span>

## <span style="color:#2980b9">**5.1 Introduction & Objective**</span>
In previous sections, we primarily used **Base 2**. However, the choice of the logarithmic base ($b$) is a critical design parameter in LNS. It dictates how the available bits are distributed across the number line.

With a fixed number of bits, we face a fundamental trade-off:
*   **Dynamic Range:** How wide is the span between the smallest and largest representable numbers?
*   **Precision:** How close are two adjacent representable numbers? (Quantization Error)

## <span style="color:#2980b9">**5.2 Mathematical Formulation**</span>
Given a fixed total width $W=8$ and fractional bits $f=4$:

1.  **Dynamic Range:**
    The ratio between the maximum and minimum representable values is approximately:
    $$ \text{Dynamic Range} \approx b^{(2^W \cdot 2^{-f})} = b^{2^{W-f}} $$
    *This shows that Dynamic Range grows **exponentially** with the base.*

2.  **Quantization Step (Relative Precision):**
    The relative distance between two consecutive LNS values ($X_i$ and $X_{i+1}$) is constant:
    $$ \text{Step Ratio} = \frac{X_{i+1}}{X_i} = b^{2^{-f}} $$
    The **maximum relative error** is approximately half of this step size:
    $$ \text{Max Error} \approx \frac{b^{2^{-f}} - 1}{2} $$
    *This shows that Error grows **linearly** (for small $f$) with the base.*

## <span style="color:#e67e22">**5.3 Experimental Setup**</span>
We will analyze four different bases to visualize this trade-off:
*   **Bases:** $\{ \sqrt{2}, 2, 4, 10 \}$
*   **Fixed Parameters:** $W=8$ bits, $f=4$ bits.
*   **Method:** For each base, we calculate the representable range, generate uniform samples, and compute the mean relative conversion error.

</div>

```python
# =============================================================================
# In this section, we analyze how the choice of logarithm base affects:
# 1. The range of representable real numbers
# 2. The conversion error (quantization error)
#
# Parameters: W=8 (total bits), f=4 (fractional bits)
# Bases to compare: √2, 2, 4, 10
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt
from typing import Tuple, Dict, List

# -----------------------------------------------------------------------------
# LNS Core Functions (from previous sections)
# -----------------------------------------------------------------------------

def get_signed_mantissa(m: int, W: int) -> int:
    """Convert two's complement mantissa to signed integer."""
    if m >= (1 << (W - 1)):
        return m - (1 << W)
    return m


def to_twos_complement(val: int, W: int) -> int:
    """Convert signed integer to two's complement representation."""
    if val < 0:
        return (1 << W) + val
    return val


def real_to_lns(x: float, base: float, W: int, f: int) -> Tuple[int, int, int]:
    """
    Convert real number to LNS representation.

    Args:
        x: Real number to convert
        base: Logarithm base
        W: Total bit width for mantissa
        f: Number of fractional bits

    Returns:
        Tuple (z, s, m) - zero flag, sign, mantissa
    """
    # Handle zero case
    if x == 0:
        return (1, 0, 0)

    # Determine sign
    s = 0 if x > 0 else 1

    # Compute logarithm of absolute value
    log_val = np.log(abs(x)) / np.log(base)

    # Convert to fixed-point representation
    scale = 2 ** f
    m_int = int(np.round(log_val * scale))

    # Clamp to W-bit two's complement range
    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_int = max(min_val, min(max_val, m_int))

    # Convert to two's complement if negative
    if m_int < 0:
        m = (1 << W) + m_int
    else:
        m = m_int

    return (0, s, m)


def lns_to_real(z: int, s: int, m: int, base: float, W: int, f: int) -> float:
    """
    Convert LNS representation back to real number.

    Args:
        z: Zero flag
        s: Sign bit
        m: Mantissa in two's complement
        base: Logarithm base
        W: Total bit width
        f: Fractional bits

    Returns:
        Real number value
    """
    if z == 1:
        return 0.0

    # Convert from two's complement
    m_signed = get_signed_mantissa(m, W)

    # Convert fixed-point to real log value
    log_val = m_signed / (2 ** f)

    # Compute absolute value and apply sign
    abs_val = base ** log_val
    return -abs_val if s == 1 else abs_val


def get_lns_range(base: float, W: int, f: int) -> Tuple[float, float]:
    """
    Calculate the range of positive real numbers representable in LNS.

    Args:
        base: Logarithm base
        W: Total bit width
        f: Fractional bits

    Returns:
        Tuple (min_positive, max_positive)
    """
    # Two's complement range for W bits
    min_log = -(2 ** (W - 1)) / (2 ** f)
    max_log = ((2 ** (W - 1)) - 1) / (2 ** f)

    # Convert to real values
    min_val = base ** min_log
    max_val = base ** max_log

    return (min_val, max_val)


def calculate_errors(original: float, converted: float) -> Tuple[float, float]:
    """Calculate absolute and relative errors."""
    abs_error = abs(converted - original)

    if original == 0:
        rel_error = 0.0 if converted == 0 else float('inf')
    else:
        rel_error = abs_error / abs(original)

    return (abs_error, rel_error)
```

### MAIN ANALYSIS

## <span style="color: #009688;">Section 6: Neural Network Convolution in LNS Domain</span>

### **6.1 Motivation**
The ultimate goal of studying LNS is to accelerate Neural Networks. The **Convolution** operation involves calculating the dot product of an Input Tensor ($I$) and a Filter/Weight Tensor ($F$):
$$ Y = \sum_{i} (I_i \cdot F_i) $$

In LNS, this becomes computationally efficient:
1.  **Multiplication:** Becomes exact integer addition ($ \log(A \cdot B) = \log A + \log B $).
2.  **Accumulation:** Requires approximate LNS addition ($\Sigma$).

### **6.2 Experiment Description**
We simulate a single convolutional layer operation to measure the real-world impact of LNS arithmetic errors.

*   **Tensors:** Input ($7 \times 7 \times 3$) and Filter ($7 \times 7 \times 3$).
*   **Total Operations:** 147 Multiplications + 146 Additions.
*   **Ranges:** Inputs $\in [0, 4]$, Weights $\in [-4, 4]$.
*   **LNS Config:** Base 2, $W=8, f=4$.

We compare the output of the standard **Real Domain Convolution** against LNS implementations using:
1.  **Exact $\Delta$**
2.  **LUT Approximation (Size 32)**
3.  **Bit-Shift Approximation**

</div>

```python
# In this section, we:
# 1. Create a simple neural network convolution layer
# 2. Input and filter dimensions: 7×7×3
# 3. Perform convolution in real domain (reference)
# 4. Perform convolution in LNS domain using all three addition methods
# 5. Compare results and analyze errors
#
# Parameters: base=2, W=8, f=4
# Input range: [0, 4]
# Filter range: [-4, 4]
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt
from typing import Tuple, Dict, List

# -----------------------------------------------------------------------------
# LNS Core Functions (from previous sections)
# -----------------------------------------------------------------------------

def get_signed_mantissa(m: int, W: int) -> int:
    """Convert two's complement mantissa to signed integer."""
    if m >= (1 << (W - 1)):
        return m - (1 << W)
    return m


def to_twos_complement(val: int, W: int) -> int:
    """Convert signed integer to two's complement representation."""
    if val < 0:
        return (1 << W) + val
    return val


def real_to_lns(x: float, base: float, W: int, f: int) -> Tuple[int, int, int]:
    """
    Convert real number to LNS representation.

    Returns:
        Tuple (z, s, m) - zero flag, sign, mantissa
    """
    if x == 0:
        return (1, 0, 0)

    s = 0 if x > 0 else 1
    log_val = np.log(abs(x)) / np.log(base)

    scale = 2 ** f
    m_int = int(np.round(log_val * scale))

    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_int = max(min_val, min(max_val, m_int))

    if m_int < 0:
        m = (1 << W) + m_int
    else:
        m = m_int

    return (0, s, m)


def lns_to_real(z: int, s: int, m: int, base: float, W: int, f: int) -> float:
    """Convert LNS representation back to real number."""
    if z == 1:
        return 0.0

    m_signed = get_signed_mantissa(m, W)
    log_val = m_signed / (2 ** f)
    abs_val = base ** log_val

    return -abs_val if s == 1 else abs_val


# -----------------------------------------------------------------------------
# LNS Multiplication (simple in LNS: just add mantissas)
# -----------------------------------------------------------------------------

def lns_multiply(lns1: Tuple[int, int, int], lns2: Tuple[int, int, int],
                 base: float, W: int, f: int) -> Tuple[int, int, int]:
    """
    Multiply two LNS numbers.

    In LNS domain: multiplication = addition of mantissas
    log(a * b) = log(a) + log(b)

    Args:
        lns1, lns2: LNS representations (z, s, m)
        base, W, f: LNS parameters

    Returns:
        LNS representation of the product
    """
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    # If either operand is zero, result is zero
    if z1 == 1 or z2 == 1:
        return (1, 0, 0)

    # Result sign: XOR of signs
    result_sign = s1 ^ s2

    # Add mantissas (in signed form)
    m1_signed = get_signed_mantissa(m1, W)
    m2_signed = get_signed_mantissa(m2, W)

    m_result = m1_signed + m2_signed

    # Clamp to valid range
    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_result = max(min_val, min(max_val, m_result))

    # Convert to two's complement
    m_result_tc = to_twos_complement(m_result, W)

    return (0, result_sign, m_result_tc)


# -----------------------------------------------------------------------------
# LNS Addition Functions (all three methods)
# -----------------------------------------------------------------------------

def lns_add_exact(lns1: Tuple[int, int, int], lns2: Tuple[int, int, int],
                  base: float, W: int, f: int) -> Tuple[int, int, int]:
    """LNS addition with exact Δ computation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    # Handle zero cases
    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    # Get signed mantissas and convert to log values
    m1_signed = get_signed_mantissa(m1, W)
    m2_signed = get_signed_mantissa(m2, W)
    m1_real = m1_signed / (2 ** f)
    m2_real = m2_signed / (2 ** f)

    # Find max and compute d
    if m1_real >= m2_real:
        m_max, s_max = m1_real, s1
        s_min = s2
    else:
        m_max, s_max = m2_real, s2
        s_min = s1

    d = abs(m1_real - m2_real)

    # Compute delta based on signs
    if s_max == s_min:
        # Same signs: addition
        delta = np.log(1 + base ** (-d)) / np.log(base)
        result_sign = s_max
    else:
        # Different signs: subtraction
        inner = 1 - base ** (-d)
        if inner <= 0:
            return (1, 0, 0)
        delta = np.log(inner) / np.log(base)
        result_sign = s_max

    # Compute result mantissa
    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    # Clamp
    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))


class LNS_LUT:
    """Lookup table for LNS addition delta values."""

    def __init__(self, base: float, W: int, f: int, lut_size: int = 32):
        self.base = base
        self.W = W
        self.f = f
        self.lut_size = lut_size

        # Maximum d value to cover
        self.d_max = W - f + 4
        self.d_step = self.d_max / (lut_size - 1) if lut_size > 1 else 1

        # Generate LUT values
        self.d_values = np.linspace(0, self.d_max, lut_size)

        # Δ+ = log_base(1 + 2^(-d))
        self.delta_plus = np.log(1 + 2 ** (-self.d_values)) / np.log(base)

        # Δ- = log_base(1 - 2^(-d))
        with np.errstate(divide='ignore', invalid='ignore'):
            self.delta_minus = np.log(1 - 2 ** (-self.d_values)) / np.log(base)
        self.delta_minus[0] = -np.inf

    def lookup_delta_plus(self, d: float) -> float:
        d = max(0, min(d, self.d_max))
        idx = int(round(d / self.d_step)) if self.d_step > 0 else 0
        idx = max(0, min(idx, self.lut_size - 1))
        return self.delta_plus[idx]

    def lookup_delta_minus(self, d: float) -> float:
        d = max(0, min(d, self.d_max))
        idx = int(round(d / self.d_step)) if self.d_step > 0 else 0
        idx = max(0, min(idx, self.lut_size - 1))
        return self.delta_minus[idx]


def lns_add_lut(lns1: Tuple[int, int, int], lns2: Tuple[int, int, int],
                base: float, W: int, f: int, lut: LNS_LUT) -> Tuple[int, int, int]:
    """LNS addition using LUT-based Δ approximation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    m1_signed = get_signed_mantissa(m1, W)
    m2_signed = get_signed_mantissa(m2, W)
    m1_real = m1_signed / (2 ** f)
    m2_real = m2_signed / (2 ** f)

    if m1_real >= m2_real:
        m_max, s_max = m1_real, s1
        s_min = s2
    else:
        m_max, s_max = m2_real, s2
        s_min = s1

    d = abs(m1_real - m2_real)

    if s_max == s_min:
        delta = lut.lookup_delta_plus(d)
        result_sign = s_max
    else:
        delta = lut.lookup_delta_minus(d)
        if np.isinf(delta) and delta < 0:
            return (1, 0, 0)
        result_sign = s_max

    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))


def lns_add_shift(lns1: Tuple[int, int, int], lns2: Tuple[int, int, int],
                  base: float, W: int, f: int) -> Tuple[int, int, int]:
    """LNS addition using bit-shift approximation."""
    z1, s1, m1 = lns1
    z2, s2, m2 = lns2

    if z1 == 1 and z2 == 1:
        return (1, 0, 0)
    if z1 == 1:
        return lns2
    if z2 == 1:
        return lns1

    m1_signed = get_signed_mantissa(m1, W)
    m2_signed = get_signed_mantissa(m2, W)
    m1_real = m1_signed / (2 ** f)
    m2_real = m2_signed / (2 ** f)

    if m1_real >= m2_real:
        m_max, s_max = m1_real, s1
        s_min = s2
    else:
        m_max, s_max = m2_real, s2
        s_min = s1

    d = abs(m1_real - m2_real)
    d_int = int(np.floor(d))

    if s_max == s_min:
        if d_int >= W:
            delta = 0
        else:
            delta = np.log(1 + 2**(-max(d_int, 0))) / np.log(base) if d_int >= 0 else 1/np.log(base)
        result_sign = s_max
    else:
        if d < 0.1:
            return (1, 0, 0)
        if d_int >= W:
            delta = 0
        else:
            inner = 1 - 2**(-max(d_int, 1)) if d_int > 0 else 0
            if inner <= 0:
                return (1, 0, 0)
            delta = np.log(inner) / np.log(base)
        result_sign = s_max

    m_result_real = m_max + delta
    m_result_fp = int(np.round(m_result_real * (2 ** f)))

    min_val = -(2 ** (W - 1))
    max_val = (2 ** (W - 1)) - 1
    m_result_fp = max(min_val, min(max_val, m_result_fp))

    return (0, result_sign, to_twos_complement(m_result_fp, W))


# =============================================================================
# CONVOLUTION FUNCTIONS
# =============================================================================

def convolution_real(input_tensor: np.ndarray, filter_tensor: np.ndarray) -> float:
    """
    Perform convolution in real domain.

    For 7×7×3 input and 7×7×3 filter, output is a scalar.
    Convolution = sum of element-wise products.

    Args:
        input_tensor: Input array of shape (7, 7, 3)
        filter_tensor: Filter array of shape (7, 7, 3)

    Returns:
        Scalar result of convolution
    """
    return np.sum(input_tensor * filter_tensor)


def convolution_lns(input_tensor: np.ndarray, filter_tensor: np.ndarray,
                    base: float, W: int, f: int,
                    add_method: str = 'exact', lut: LNS_LUT = None) -> float:
    """
    Perform convolution in LNS domain.

    Steps:
    1. Convert all inputs and weights to LNS
    2. Perform element-wise multiplication in LNS (add mantissas)
    3. Sum all products using specified addition method
    4. Convert final result back to real

    Args:
        input_tensor: Input array
        filter_tensor: Filter array
        base, W, f: LNS parameters
        add_method: 'exact', 'lut', or 'shift'
        lut: LUT object (required if add_method='lut')

    Returns:
        Scalar result of convolution
    """
    # Flatten tensors for easier processing
    inputs_flat = input_tensor.flatten()
    filters_flat = filter_tensor.flatten()

    # Convert all values to LNS
    inputs_lns = [real_to_lns(x, base, W, f) for x in inputs_flat]
    filters_lns = [real_to_lns(x, base, W, f) for x in filters_flat]

    # Perform element-wise multiplication in LNS
    products_lns = []
    for inp, flt in zip(inputs_lns, filters_lns):
        prod = lns_multiply(inp, flt, base, W, f)
        products_lns.append(prod)

    # Sum all products using specified addition method
    # Start with first product
    accumulator = products_lns[0]

    for i in range(1, len(products_lns)):
        if add_method == 'exact':
            accumulator = lns_add_exact(accumulator, products_lns[i], base, W, f)
        elif add_method == 'lut':
            accumulator = lns_add_lut(accumulator, products_lns[i], base, W, f, lut)
        elif add_method == 'shift':
            accumulator = lns_add_shift(accumulator, products_lns[i], base, W, f)
        else:
            raise ValueError(f"Unknown addition method: {add_method}")

    # Convert final result back to real
    result = lns_to_real(*accumulator, base, W, f)

    return result


# =============================================================================
# MAIN ANALYSIS
# =============================================================================

print("=" * 90)
print("SECTION 6: NEURAL NETWORK CONVOLUTION IN LNS DOMAIN")
print("=" * 90)

# Parameters as specified in the problem
base = 2
W = 8
f = 4
lut_size = 32

print(f"\nLNS Parameters: base={base}, W={W}, f={f}, LUT size={lut_size}")

# Input and filter dimensions
input_shape = (7, 7, 3)
filter_shape = (7, 7, 3)
total_elements = np.prod(input_shape)

print(f"Input shape: {input_shape}")
print(f"Filter shape: {filter_shape}")
print(f"Total elements: {total_elements}")

# Generate random input and filter
np.random.seed(42)  # For reproducibility

# Input range: [0, 4] (positive values only as specified)
input_tensor = np.random.uniform(0, 4, input_shape)

# Filter range: [-4, 4]
filter_tensor = np.random.uniform(-4, 4, filter_shape)

print(f"\nInput tensor range: [{input_tensor.min():.4f}, {input_tensor.max():.4f}]")
print(f"Filter tensor range: [{filter_tensor.min():.4f}, {filter_tensor.max():.4f}]")

# Create LUT for LUT-based addition
lut = LNS_LUT(base, W, f, lut_size)

# -----------------------------------------------------------------------------
# Perform Convolution with All Methods
# -----------------------------------------------------------------------------

print(f"\n{'=' * 70}")
print("CONVOLUTION RESULTS")
print(f"{'=' * 70}")

# Real domain convolution (reference)
result_real = convolution_real(input_tensor, filter_tensor)
print(f"\n1. Real Domain Convolution (Reference):")
print(f"   Result: {result_real:.6f}")

# LNS domain with exact addition
result_lns_exact = convolution_lns(input_tensor, filter_tensor, base, W, f,
                                    add_method='exact')
error_exact = abs(result_lns_exact - result_real)
rel_error_exact = error_exact / abs(result_real) * 100 if result_real != 0 else 0

print(f"\n2. LNS Domain - Exact Δ Computation:")
print(f"   Result: {result_lns_exact:.6f}")
print(f"   Absolute Error: {error_exact:.6f}")
print(f"   Relative Error: {rel_error_exact:.4f}%")

# LNS domain with LUT-based addition
result_lns_lut = convolution_lns(input_tensor, filter_tensor, base, W, f,
                                  add_method='lut', lut=lut)
error_lut = abs(result_lns_lut - result_real)
rel_error_lut = error_lut / abs(result_real) * 100 if result_real != 0 else 0

print(f"\n3. LNS Domain - LUT Approximation (size={lut_size}):")
print(f"   Result: {result_lns_lut:.6f}")
print(f"   Absolute Error: {error_lut:.6f}")
print(f"   Relative Error: {rel_error_lut:.4f}%")

# LNS domain with bit-shift addition
result_lns_shift = convolution_lns(input_tensor, filter_tensor, base, W, f,
                                    add_method='shift')
error_shift = abs(result_lns_shift - result_real)
rel_error_shift = error_shift / abs(result_real) * 100 if result_real != 0 else 0

print(f"\n4. LNS Domain - Bit-Shift Approximation:")
print(f"   Result: {result_lns_shift:.6f}")
print(f"   Absolute Error: {error_shift:.6f}")
print(f"   Relative Error: {rel_error_shift:.4f}%")

# -----------------------------------------------------------------------------
# Summary Table
# -----------------------------------------------------------------------------

print(f"\n{'=' * 70}")
print("SUMMARY TABLE")
print(f"{'=' * 70}")

print(f"\n{'Method':<30} {'Result':<15} {'Abs Error':<15} {'Rel Error':<15}")
print(f"{'─' * 75}")
print(f"{'Real Domain (Reference)':<30} {result_real:<15.6f} {'—':<15} {'—':<15}")
print(f"{'LNS - Exact Δ':<30} {result_lns_exact:<15.6f} {error_exact:<15.6f} {rel_error_exact:<14.4f}%")
print(f"{'LNS - LUT (32 entries)':<30} {result_lns_lut:<15.6f} {error_lut:<15.6f} {rel_error_lut:<14.4f}%")
print(f"{'LNS - Bit-Shift':<30} {result_lns_shift:<15.6f} {error_shift:<15.6f} {rel_error_shift:<14.4f}%")

# -----------------------------------------------------------------------------
# Multiple Trials Analysis
# -----------------------------------------------------------------------------

print(f"\n{'=' * 70}")
print("MULTIPLE TRIALS ANALYSIS (10 random initializations)")
print(f"{'=' * 70}")

num_trials = 10
errors_exact = []
errors_lut = []
errors_shift = []

for trial in range(num_trials):
    # Generate new random tensors
    np.random.seed(trial * 100)
    inp = np.random.uniform(0, 4, input_shape)
    flt = np.random.uniform(-4, 4, filter_shape)

    # Compute convolutions
    ref = convolution_real(inp, flt)
    lns_exact = convolution_lns(inp, flt, base, W, f, 'exact')
    lns_lut = convolution_lns(inp, flt, base, W, f, 'lut', lut)
    lns_shift = convolution_lns(inp, flt, base, W, f, 'shift')

    # Calculate relative errors
    if ref != 0:
        errors_exact.append(abs(lns_exact - ref) / abs(ref) * 100)
        errors_lut.append(abs(lns_lut - ref) / abs(ref) * 100)
        errors_shift.append(abs(lns_shift - ref) / abs(ref) * 100)

errors_exact = np.array(errors_exact)
errors_lut = np.array(errors_lut)
errors_shift = np.array(errors_shift)

print(f"\n{'Method':<25} {'Mean Err%':<15} {'Max Err%':<15} {'Std Err%':<15}")
print(f"{'─' * 70}")
print(f"{'LNS - Exact Δ':<25} {np.mean(errors_exact):<15.4f} {np.max(errors_exact):<15.4f} {np.std(errors_exact):<15.4f}")
print(f"{'LNS - LUT':<25} {np.mean(errors_lut):<15.4f} {np.max(errors_lut):<15.4f} {np.std(errors_lut):<15.4f}")
print(f"{'LNS - Bit-Shift':<25} {np.mean(errors_shift):<15.4f} {np.max(errors_shift):<15.4f} {np.std(errors_shift):<15.4f}")

# -----------------------------------------------------------------------------
# Visualization
# -----------------------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Section 6: Neural Network Convolution in LNS Domain', fontsize=14, fontweight='bold')

# Plot 1: Results comparison bar chart
ax1 = axes[0]
methods = ['Real\n(Reference)', 'LNS\nExact', 'LNS\nLUT', 'LNS\nShift']
results = [result_real, result_lns_exact, result_lns_lut, result_lns_shift]
colors = ['green', 'blue', 'orange', 'red']
bars = ax1.bar(methods, results, color=colors, edgecolor='black', alpha=0.7)
ax1.set_ylabel('Convolution Result', fontsize=11)
ax1.set_title('Convolution Results Comparison', fontsize=12)
ax1.axhline(y=result_real, color='green', linestyle='--', alpha=0.5, label='Reference')
ax1.grid(True, alpha=0.3, axis='y')

# Add value labels
for bar, val in zip(bars, results):
    ax1.annotate(f'{val:.2f}', xy=(bar.get_x() + bar.get_width()/2, bar.get_height()),
                 xytext=(0, 5), textcoords='offset points', ha='center', fontsize=9)

# Plot 2: Relative errors bar chart
ax2 = axes[1]
error_methods = ['Exact Δ', 'LUT', 'Bit-Shift']
rel_errors = [rel_error_exact, rel_error_lut, rel_error_shift]
error_colors = ['blue', 'orange', 'red']
bars2 = ax2.bar(error_methods, rel_errors, color=error_colors, edgecolor='black', alpha=0.7)
ax2.set_ylabel('Relative Error (%)', fontsize=11)
ax2.set_title('Relative Error by Method', fontsize=12)
ax2.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars2, rel_errors):
    ax2.annotate(f'{val:.2f}%', xy=(bar.get_x() + bar.get_width()/2, bar.get_height()),
                 xytext=(0, 5), textcoords='offset points', ha='center', fontsize=10)

# Plot 3: Multiple trials error distribution
ax3 = axes[2]
positions = [1, 2, 3]
bp = ax3.boxplot([errors_exact, errors_lut, errors_shift], positions=positions,
                  patch_artist=True, widths=0.6)

colors_box = ['lightblue', 'lightyellow', 'lightcoral']
for patch, color in zip(bp['boxes'], colors_box):
    patch.set_facecolor(color)

ax3.set_xticklabels(['Exact Δ', 'LUT', 'Bit-Shift'])
ax3.set_ylabel('Relative Error (%)', fontsize=11)
ax3.set_title(f'Error Distribution ({num_trials} Trials)', fontsize=12)
ax3.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('section6_convolution.png', dpi=150, bbox_inches='tight')
plt.show()

# -----------------------------------------------------------------------------
# Error Source Analysis
# -----------------------------------------------------------------------------

print(f"\n{'=' * 70}")
print("ERROR SOURCE ANALYSIS")
print(f"{'=' * 70}")

# Analyze where errors come from
print("\n1. CONVERSION ERROR:")
print("   Converting real numbers to LNS introduces quantization error.")

# Sample conversion error analysis
sample_values = [0.5, 1.0, 2.0, 3.5]
print(f"\n   Sample conversion errors:")
print(f"   {'Original':<12} {'Converted':<12} {'Rel Error%':<12}")
for val in sample_values:
    lns_rep = real_to_lns(val, base, W, f)
    converted = lns_to_real(*lns_rep, base, W, f)
    err = abs(converted - val) / val * 100 if val != 0 else 0
    print(f"   {val:<12.4f} {converted:<12.4f} {err:<12.4f}")

print("\n2. MULTIPLICATION ERROR:")
print("   LNS multiplication is exact (just addition of mantissas).")
print("   However, result may overflow/underflow the representable range.")

print("\n3. ADDITION ERROR (Main Source):")
print("   The Δ approximation in addition accumulates over many operations.")
print(f"   Total additions performed: {total_elements - 1}")

print("\n4. ACCUMULATION EFFECT:")
print("   Errors compound through the accumulation chain.")
print("   Later additions may have larger errors due to accumulated magnitude changes.")

# -----------------------------------------------------------------------------
# Detailed Analysis
# -----------------------------------------------------------------------------

print(f"\n{'=' * 90}")
print("DETAILED ANALYSIS AND CONCLUSIONS")
print(f"{'=' * 90}")

analysis = """
1. CONVOLUTION IN LNS - KEY OBSERVATIONS:
   ────────────────────────────────────────
   • The convolution operation involves 147 multiplications and 146 additions
   • Multiplications in LNS are EXACT (just mantissa addition)
   • Additions introduce approximation error through the Δ term
   • Errors accumulate throughout the 146 sequential additions

2. ERROR BREAKDOWN BY METHOD:
   ───────────────────────────
   • Exact Δ Method:
     - Only source of error is fixed-point quantization
     - Provides the best accuracy among LNS methods
     - Still has ~X% error due to quantization effects

   • LUT Method:
     - Additional error from discrete LUT approximation
     - Error depends on LUT size and distribution of d values
     - Good balance between accuracy and computational cost

   • Bit-Shift Method:
     - Highest error due to coarse approximation
     - Uses floor(d) instead of exact d
     - Most efficient but least accurate

3. COMPARISON WITH REAL ARITHMETIC:
   ──────────────────────────────────
   • Real arithmetic provides exact result (reference)
   • LNS methods trade accuracy for:
     - Simplified multiplication (addition instead of multiply)
     - Potential hardware efficiency (logarithmic operations)
     - Reduced dynamic range issues

4. PRACTICAL IMPLICATIONS FOR NEURAL NETWORKS:
   ────────────────────────────────────────────
   • Error magnitude is significant for a single layer
   • Deep networks would accumulate more errors across layers
   • However, neural networks are often error-tolerant:
     - Training can adapt to systematic biases
     - Small errors may not affect classification accuracy

   • Trade-offs to consider:
     - Exact Δ: Most accurate but computationally expensive
     - LUT: Good accuracy with moderate memory cost
     - Bit-Shift: Fastest but may need error compensation

5. RECOMMENDATIONS:
   ─────────────────
   • For inference in resource-constrained devices:
     - LUT method with 32+ entries provides good accuracy

   • For training:
     - Consider exact method or hybrid approaches

   • For very deep networks:
     - May need periodic conversion back to real domain
     - Or use higher precision LNS parameters
"""

print(analysis)

# -----------------------------------------------------------------------------
# Final Summary
# -----------------------------------------------------------------------------

print(f"\n{'=' * 90}")
print("FINAL SUMMARY - SECTION 6")
print(f"{'=' * 90}")

print(f"""
┌─────────────────────────────────────────────────────────────────────────────┐
│                    CONVOLUTION RESULTS SUMMARY                              │
├─────────────────────────────────────────────────────────────────────────────┤
│  Configuration: Input & Filter = 7×7×3, base=2, W=8, f=4                    │
├─────────────────────────────────────────────────────────────────────────────┤
│  Method              │  Result        │  Relative Error                     │
├─────────────────────────────────────────────────────────────────────────────┤
│  Real Domain         │  {result_real:>12.4f}  │  (Reference)                        │
│  LNS - Exact Δ       │  {result_lns_exact:>12.4f}  │  {rel_error_exact:>6.2f}%                            │
│  LNS - LUT (32)      │  {result_lns_lut:>12.4f}  │  {rel_error_lut:>6.2f}%                            │
│  LNS - Bit-Shift     │  {result_lns_shift:>12.4f}  │  {rel_error_shift:>6.2f}%                            │
└─────────────────────────────────────────────────────────────────────────────┘

KEY TAKEAWAY: LNS arithmetic can approximate neural network convolution with
reasonable accuracy. The LUT method offers the best trade-off between accuracy
and computational efficiency for hardware implementation.
""")

print(f"\n✓ Section 6 analysis complete!")
print(f"  Plot saved: 'section6_convolution.png'")
```

## <span style="color: #2e7d32;">6.3 Final Analysis & Conclusion</span>

This experiment reveals the challenges of deploying LNS in Deep Learning inference.

### **Analysis of Results**

1.  **Error Accumulation:**
    *   Unlike previous sections with single operations, convolution involves **146 sequential additions**.
    *   Even small errors in individual additions accumulate (drift) significantly over the summation chain.
    *   The **Exact LNS** method shows ~13% mean error solely due to quantization and accumulation drift.

2.  **Performance of Approximations:**
    *   **LUT Method:** Shows a mean error of **~37%**. While high, this follows the trend of the exact method and maintains the correct sign and order of magnitude.
    *   **Bit-Shift Method:** Shows a high mean error (**~52%**) and sometimes produces completely incorrect values (e.g., wrong sign). This confirms that Bit-Shift is **unsuitable** for deep networks where weights can be negative (subtraction), as the linear approximation fails in the non-linear subtraction regions.

3.  **Recommendation for Hardware Implementation:**
    *   For **Inference:** A **LUT-based** LNS accelerator is feasible. The error can be mitigated by training the network *with* the LNS constraints (Quantization Aware Training).
    *   **Accumulator Precision:** To reduce the accumulation error, the internal accumulator should have a **higher bit-width** (e.g., 16 bits) before converting back to 8-bit output.

### **✅ Final Project Conclusion**
This study demonstrates that **Logarithmic Number Systems** offer a powerful alternative to floating-point arithmetic by reducing complex multiplications to simple additions. However, the non-linear nature of LNS addition introduces errors that must be carefully managed via **Base selection**, **LUT sizing**, and **Accumulator precision**.
