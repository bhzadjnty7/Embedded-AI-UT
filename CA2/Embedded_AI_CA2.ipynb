## 📚 Overview

In this section, we implement the **forward conversion** process to transform integers from the standard decimal representation into the **Residue Number System (RNS)** domain. RNS is a non-weighted number system where an integer is represented as a set of remainders with respect to a collection of pairwise coprime moduli.

## 🎯 Objectives

1. **Implement conversion functions** for both signed and unsigned integers
2. **Verify uniqueness** of RNS representations within the dynamic range
3. **Demonstrate conversion** using moduli set `{2, 3, 5}`
4. **Validate** that no two distinct integers share the same RNS representation

</div>

<div style="background-color: #fff7ed; padding: 20px; border-radius: 8px; margin: 20px 0; border: 2px solid #fb923c;">

## 🔬 Mathematical Foundation

### Dynamic Range
For a set of moduli **M = {m₁, m₂, ..., mₙ}**, the dynamic range is defined as:

<div style="text-align: center; padding: 15px; background-color: white; border-radius: 8px; margin: 10px 0; font-size: 18px;">
<strong>M = m₁ × m₂ × ... × mₙ</strong>
</div>

This represents the total count of unique numbers that can be represented in the RNS domain.

### Forward Conversion Formulas

#### Unsigned Representation
For an unsigned integer **x** where **0 ≤ x < M**:

<div style="text-align: center; padding: 15px; background-color: white; border-radius: 8px; margin: 10px 0;">
<strong>RNS(x) = [x mod m₁, x mod m₂, ..., x mod mₙ]</strong>
</div>

**Valid Range:** `[0, M-1]`

#### Signed Representation
For a signed integer **x** where **-⌊M/2⌋ ≤ x ≤ ⌊(M-1)/2⌋**:

<div style="text-align: center; padding: 15px; background-color: white; border-radius: 8px; margin: 10px 0;">
<strong>RNS(x) = [(x mod M) mod m₁, (x mod M) mod m₂, ..., (x mod M) mod mₙ]</strong>
</div>

For negative numbers, we use the equivalence: **x < 0 → x + M** (modular arithmetic)

**Valid Range:** `[-⌊M/2⌋, ⌊(M-1)/2⌋]`

</div>

<div style="background-color: #ecfdf5; padding: 20px; border-radius: 8px; margin: 20px 0; border: 2px solid #10b981;">

## 📝 Example: Moduli Set {2, 3, 5}

**Dynamic Range:** M = 2 × 3 × 5 = **30**

**Unsigned Range:** [0, 29] → 30 unique representations  
**Signed Range:** [-15, 14] → 30 unique representations

### Conversion Example (Unsigned)
- **x = 23** → RNS = [23 mod 2, 23 mod 3, 23 mod 5] = **[1, 2, 3]**
- **x = 15** → RNS = [15 mod 2, 15 mod 3, 15 mod 5] = **[1, 0, 0]**

### Conversion Example (Signed)
- **x = -7** → x + M = -7 + 30 = 23 → RNS = **[1, 2, 3]**
- **x = 5** → RNS = [5 mod 2, 5 mod 3, 5 mod 5] = **[1, 2, 0]**

</div>

<div style="background-color: #fef3c7; padding: 15px; border-radius: 8px; margin: 20px 0; border-left: 5px solid #f59e0b;">

## ⚠️ Important Notes

- **Coprime Requirement:** All moduli must be pairwise coprime (gcd(mᵢ, mⱼ) = 1 for i ≠ j)
- **Range Limitation:** Numbers outside the valid range cannot be uniquely represented
- **Signed Convention:** Negative numbers are represented using their positive modular equivalent
- **Overflow:** Operations may produce results outside the dynamic range requiring careful handling

# **Part 1:  Forward Conversion**

## Converting Signed and Unsigned Integers to RNS Representation

```python
import numpy as np
from typing import List, Tuple

def compute_dynamic_range(moduli: List[int]) -> int:
    """
    Compute the dynamic range M = product of all moduli.
    This represents the total range of representable numbers in RNS.
    """
    M = 1
    for m in moduli:
        M *= m
    return M

def forward_conversion_unsigned(x: int, moduli: List[int]) -> List[int]:
    """
    Convert an unsigned integer to its RNS representation.

    Args:
        x: Non-negative integer to convert
        moduli: List of pairwise coprime moduli

    Returns:
        List of residues [x mod m1, x mod m2, ..., x mod mn]
    """
    return [x % m for m in moduli]

def forward_conversion_signed(x: int, moduli: List[int]) -> List[int]:
    """
    Convert a signed integer to its RNS representation.
    For negative numbers, we use the property: -x mod m = m - (x mod m)

    Args:
        x: Signed integer to convert
        moduli: List of pairwise coprime moduli

    Returns:
        List of residues representing the signed number in RNS
    """
    M = compute_dynamic_range(moduli)

    # For signed representation, valid range is [-M//2, M//2) or [-M//2, M//2-1]
    if x < 0:
        # Convert negative number: add M to bring it to positive equivalent
        x = x + M

    return [x % m for m in moduli]

def get_unsigned_range(moduli: List[int]) -> Tuple[int, int]:
    """
    Get the range of unsigned numbers representable with given moduli.
    Range: [0, M-1] where M = product of moduli
    """
    M = compute_dynamic_range(moduli)
    return (0, M - 1)

def get_signed_range(moduli: List[int]) -> Tuple[int, int]:
    """
    Get the range of signed numbers representable with given moduli.
    Range: [-M//2, M//2 - 1] for even M, or [-M//2, M//2] for odd M
    """
    M = compute_dynamic_range(moduli)
    return (-(M // 2), (M - 1) // 2)

def verify_uniqueness(moduli: List[int], signed: bool = False) -> bool:
    """
    Verify that all numbers in the valid range have unique RNS representations.

    Args:
        moduli: List of pairwise coprime moduli
        signed: If True, check signed range; otherwise check unsigned range

    Returns:
        True if all representations are unique, False otherwise
    """
    if signed:
        start, end = get_signed_range(moduli)
    else:
        start, end = get_unsigned_range(moduli)

    representations = {}

    for x in range(start, end + 1):
        if signed:
            rns = forward_conversion_signed(x, moduli)
        else:
            rns = forward_conversion_unsigned(x, moduli)

        rns_tuple = tuple(rns)

        if rns_tuple in representations:
            print(f"Duplicate found! {x} and {representations[rns_tuple]} both map to {rns}")
            return False

        representations[rns_tuple] = x

    return True


# ==================== Testing with moduli {2, 3, 5} ====================

moduli = [2, 3, 5]
M = compute_dynamic_range(moduli)

print("=" * 60)
print("RNS Forward Conversion - Part 1")
print("=" * 60)
print(f"\nModuli set: {moduli}")
print(f"Dynamic range M = {' × '.join(map(str, moduli))} = {M}")

# Unsigned numbers
print("\n" + "-" * 60)
print("UNSIGNED NUMBERS")
print("-" * 60)
unsigned_start, unsigned_end = get_unsigned_range(moduli)
print(f"Valid range: [{unsigned_start}, {unsigned_end}]")
print(f"Total numbers: {unsigned_end - unsigned_start + 1}\n")

print(f"{'Number':<10} {'RNS Representation':<25} {'(mod 2, mod 3, mod 5)'}")
print("-" * 50)

for x in range(unsigned_start, unsigned_end + 1):
    rns = forward_conversion_unsigned(x, moduli)
    print(f"{x:<10} {str(rns):<25}")

# Verify uniqueness for unsigned
is_unique_unsigned = verify_uniqueness(moduli, signed=False)
print(f"\nUniqueness verification (unsigned): {'PASSED ✓' if is_unique_unsigned else 'FAILED ✗'}")

# Signed numbers
print("\n" + "-" * 60)
print("SIGNED NUMBERS")
print("-" * 60)
signed_start, signed_end = get_signed_range(moduli)
print(f"Valid range: [{signed_start}, {signed_end}]")
print(f"Total numbers: {signed_end - signed_start + 1}\n")

print(f"{'Number':<10} {'RNS Representation':<25} {'Equivalent unsigned'}")
print("-" * 50)

for x in range(signed_start, signed_end + 1):
    rns = forward_conversion_signed(x, moduli)
    # Show equivalent unsigned value for negative numbers
    equiv = x if x >= 0 else x + M
    print(f"{x:<10} {str(rns):<25} {equiv}")

# Verify uniqueness for signed
is_unique_signed = verify_uniqueness(moduli, signed=True)
print(f"\nUniqueness verification (signed): {'PASSED ✓' if is_unique_signed else 'FAILED ✗'}")

# Summary
print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Moduli: {moduli}")
print(f"Dynamic Range (M): {M}")
print(f"Unsigned range: [0, {M-1}] → {M} unique numbers")
print(f"Signed range: [{-(M//2)}, {(M-1)//2}] → {M} unique numbers")
print(f"All unsigned representations unique: {is_unique_unsigned}")
print(f"All signed representations unique: {is_unique_signed}")
```

## 📚 Theoretical Foundation

The **Chinese Remainder Theorem (CRT)** is a fundamental mathematical tool that enables us to uniquely reconstruct an integer from its residues in the Residue Number System. This theorem forms the backbone of inverse conversion in RNS arithmetic.

### 🎯 Core Concept

Given a set of pairwise coprime moduli $\{m_1, m_2, ..., m_n\}$ and their corresponding residues $\{r_1, r_2, ..., r_n\}$, CRT guarantees the existence of a unique solution $x$ in the range $[0, M-1]$ where $M = \prod_{i=1}^{n} m_i$.

</div>

---

<div style="background: #fff3cd; padding: 20px; border-left: 5px solid #ffc107; border-radius: 8px; margin: 20px 0;">

## 🔢 The CRT Formula

The reconstruction formula is elegantly expressed as:

$$x = \left(\sum_{i=1}^{n} r_i \cdot M_i \cdot y_i\right) \bmod M$$

Where the key components are defined as:

## 🔐 Computing Modular Multiplicative Inverse

The modular inverse $y_i$ is the critical component that satisfies:

$$M_i \cdot y_i \equiv 1 \pmod{m_i}$$

### Extended Euclidean Algorithm

To compute $y_i$, we use the **Extended Euclidean Algorithm (EEA)**, which finds integers $x$ and $y$ such that:

$$a \cdot x + b \cdot y = \gcd(a, b)$$

For coprime moduli (where $\gcd(M_i, m_i) = 1$), this guarantees the existence of the modular inverse.

**Algorithm Steps:**
1. Apply EEA to find $x$ such that $M_i \cdot x + m_i \cdot y = 1$
2. The coefficient $x$ is our modular inverse: $y_i = x \bmod m_i$
3. Ensure the result is positive by adding $m_i$ if necessary

1. **$M_i$ values:**
   - $M_1 = 30/2 = 15$
   - $M_2 = 30/3 = 10$
   - $M_3 = 30/5 = 6$

2. **Modular inverses $y_i$:**
   - $y_1 = 15^{-1} \bmod 2 = 1$ (since $15 \times 1 \equiv 1 \pmod{2}$)
   - $y_2 = 10^{-1} \bmod 3 = 1$ (since $10 \times 1 \equiv 1 \pmod{3}$)
   - $y_3 = 6^{-1} \bmod 5 = 1$ (since $6 \times 1 \equiv 1 \pmod{5}$)

$$x = (r_1 \cdot 15 \cdot 1 + r_2 \cdot 10 \cdot 1 + r_3 \cdot 6 \cdot 1) \bmod 30$$

**Example:** For RNS representation $[1, 2, 3]$:

### ✅ Uniqueness Guarantee

CRT guarantees that each integer in $[0, M-1]$ has a **unique** RNS representation, and vice versa. This bijection ensures:
- No two different integers map to the same RNS representation
- Perfect reconstruction from residues to original integer

### ✅ Computational Advantages

**Parallel Processing:** Each residue computation is independent, allowing simultaneous calculation across multiple processing units.

**No Carry Propagation:** Unlike traditional binary arithmetic, RNS operations don't require carry signals between different moduli channels.

### ⚠️ Computational Complexity

- **Forward Conversion:** O(n) where n is the number of moduli
- **Inverse Conversion (CRT):** O(n²) due to modular inverse computations
- **Precomputation Optimization:** Constants can be computed once and reused

## 🎓 Theoretical Insights

### Why CRT Works

The Chinese Remainder Theorem is based on the fundamental property that for **pairwise coprime moduli**, the mapping between integers and their residue representations is **bijective** (one-to-one and onto).

**Mathematical Proof Sketch:**
1. For coprime moduli, gcd(Mᵢ, mᵢ) = 1 always holds
2. This ensures modular inverses yᵢ exist and are unique
3. The linear combination Σ(rᵢ × Mᵢ × yᵢ) produces a unique value mod M
4. This value equals the original integer x

### Historical Context

The Chinese Remainder Theorem dates back to ancient Chinese mathematics (Sun Tzu's work, ~3rd century AD) and has applications in:
- **Cryptography** (RSA algorithm)
- **Error correction codes**
- **Parallel computing architectures**
- **Digital signal processing**

## 🔬 Implementation Notes

### Extended Euclidean Algorithm Details

The EEA computes gcd(a, b) and finds coefficients x, y such that:
a × x + b × y = gcd(a, b)

**Recursive Implementation:**
If a = 0: return (b, 0, 1)
Otherwise:
(gcd, x₁, y₁) = EEA(b mod a, a)
x = y₁ - ⌊b/a⌋ × x₁
y = x₁
return (gcd, x, y)
### Modular Inverse Computation

For coprime numbers (gcd = 1), the coefficient x gives us:
a × x ≡ 1 (mod m)
We ensure the result is positive: `inverse = (x mod m + m) mod m`

## Reconstructing Integers from RNS Domain using CRT

```python
# ============================================================
# Part 2: Inverse Conversion using Chinese Remainder Theorem (CRT)
# ============================================================

def extended_gcd(a: int, b: int) -> Tuple[int, int, int]:
    """
    Extended Euclidean Algorithm.
    Returns (gcd, x, y) such that a*x + b*y = gcd(a, b)
    """
    if a == 0:
        return b, 0, 1

    gcd, x1, y1 = extended_gcd(b % a, a)
    x = y1 - (b // a) * x1
    y = x1

    return gcd, x, y

def modular_inverse(a: int, m: int) -> int:
    """
    Compute modular multiplicative inverse of a modulo m.
    Returns x such that (a * x) mod m = 1
    """
    gcd, x, _ = extended_gcd(a % m, m)

    if gcd != 1:
        raise ValueError(f"Modular inverse does not exist for {a} mod {m}")

    return (x % m + m) % m

def crt_inverse_unsigned(residues: List[int], moduli: List[int]) -> int:
    """
    Convert RNS representation back to unsigned integer using CRT.

    CRT Formula: x = Σ(ri * Mi * yi) mod M
    where:
        - ri = residue for modulus mi
        - M = product of all moduli
        - Mi = M / mi
        - yi = modular inverse of Mi mod mi

    Args:
        residues: List of residues [r1, r2, ..., rn]
        moduli: List of pairwise coprime moduli [m1, m2, ..., mn]

    Returns:
        Unsigned integer x in range [0, M-1]
    """
    n = len(moduli)
    M = compute_dynamic_range(moduli)

    result = 0

    for i in range(n):
        Mi = M // moduli[i]  # Product of all moduli except moduli[i]
        yi = modular_inverse(Mi, moduli[i])  # Modular inverse of Mi mod moduli[i]

        result += residues[i] * Mi * yi

    return result % M

def crt_inverse_signed(residues: List[int], moduli: List[int]) -> int:
    """
    Convert RNS representation back to signed integer using CRT.

    For signed conversion, if the unsigned result >= M/2,
    we subtract M to get the negative equivalent.

    Args:
        residues: List of residues [r1, r2, ..., rn]
        moduli: List of pairwise coprime moduli [m1, m2, ..., mn]

    Returns:
        Signed integer x in range [-M//2, (M-1)//2]
    """
    M = compute_dynamic_range(moduli)

    # First get the unsigned result
    unsigned_result = crt_inverse_unsigned(residues, moduli)

    # Convert to signed: if result >= M/2, it represents a negative number
    if unsigned_result > (M - 1) // 2:
        return unsigned_result - M
    else:
        return unsigned_result


# ==================== Testing CRT Inverse Conversion ====================

print("=" * 70)
print("Part 2: Inverse Conversion using Chinese Remainder Theorem (CRT)")
print("=" * 70)

moduli = [2, 3, 5]
M = compute_dynamic_range(moduli)

print(f"\nModuli set: {moduli}")
print(f"Dynamic range M = {M}")

# Test Extended GCD and Modular Inverse
print("\n" + "-" * 70)
print("MODULAR INVERSE VERIFICATION")
print("-" * 70)

for i, m in enumerate(moduli):
    Mi = M // m
    yi = modular_inverse(Mi, m)
    print(f"M{i+1} = M/m{i+1} = {M}/{m} = {Mi}")
    print(f"y{i+1} = M{i+1}^(-1) mod m{i+1} = {Mi}^(-1) mod {m} = {yi}")
    print(f"Verification: {Mi} × {yi} mod {m} = {(Mi * yi) % m}")
    print()

# Unsigned numbers - CRT verification
print("-" * 70)
print("UNSIGNED NUMBERS - CRT INVERSE VERIFICATION")
print("-" * 70)
unsigned_start, unsigned_end = get_unsigned_range(moduli)
print(f"Valid range: [{unsigned_start}, {unsigned_end}]\n")

print(f"{'Original':<10} {'RNS':<20} {'CRT Result':<12} {'Match'}")
print("-" * 55)

all_match_unsigned = True
for x in range(unsigned_start, unsigned_end + 1):
    rns = forward_conversion_unsigned(x, moduli)
    recovered = crt_inverse_unsigned(rns, moduli)
    match = (x == recovered)
    all_match_unsigned = all_match_unsigned and match
    status = "✓" if match else "✗"
    print(f"{x:<10} {str(rns):<20} {recovered:<12} {status}")

print(f"\nAll unsigned conversions correct: {'PASSED ✓' if all_match_unsigned else 'FAILED ✗'}")

# Signed numbers - CRT verification
print("\n" + "-" * 70)
print("SIGNED NUMBERS - CRT INVERSE VERIFICATION")
print("-" * 70)
signed_start, signed_end = get_signed_range(moduli)
print(f"Valid range: [{signed_start}, {signed_end}]\n")

print(f"{'Original':<10} {'RNS':<20} {'CRT Result':<12} {'Match'}")
print("-" * 55)

all_match_signed = True
for x in range(signed_start, signed_end + 1):
    rns = forward_conversion_signed(x, moduli)
    recovered = crt_inverse_signed(rns, moduli)
    match = (x == recovered)
    all_match_signed = all_match_signed and match
    status = "✓" if match else "✗"
    print(f"{x:<10} {str(rns):<20} {recovered:<12} {status}")

print(f"\nAll signed conversions correct: {'PASSED ✓' if all_match_signed else 'FAILED ✗'}")

# Summary
print("\n" + "=" * 70)
print("PART 2 SUMMARY")
print("=" * 70)
print(f"Moduli: {moduli}")
print(f"Dynamic Range (M): {M}")
print(f"\nCRT Formula: x = Σ(ri × Mi × yi) mod M")
print(f"where Mi = M/mi and yi = Mi^(-1) mod mi")
print(f"\nUnsigned conversion verification: {'PASSED ✓' if all_match_unsigned else 'FAILED ✗'}")
print(f"Signed conversion verification: {'PASSED ✓' if all_match_signed else 'FAILED ✗'}")
```

## Garner’s Algorithm for RNS to Integer Conversion

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 30px; border-radius: 15px; margin: 20px 0;">
    <h1 style="color: white; text-align: center; font-size: 2.5em; margin: 0; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);">

    </h2>

<div style="background: linear-gradient(to right, #f8f9fa, #e9ecef); padding: 25px; border-left: 5px solid #667eea; border-radius: 10px; margin: 20px 0;">

## 📚 Theoretical Foundation

**Mixed Radix Conversion (MRC)** provides an alternative approach to inverse conversion in RNS. Unlike CRT which computes the result directly, MRC converts the RNS representation to a mixed-radix representation first, then reconstructs the integer.

### 🎯 Core Concept

**Garner's Algorithm** is the most efficient method for implementing MRC. It computes intermediate values called **mixed-radix digits** that uniquely represent the number in a different positional system.

</div>

---

<div style="background: #fff3cd; padding: 20px; border-left: 5px solid #ffc107; border-radius: 8px; margin: 20px 0;">

## 🔢 The MRC Formula

The reconstruction is performed in two phases:

### Phase 1: Computing Mixed-Radix Digits

$$v_0 = r_0$$

$$v_i = \left(\left(\left(r_i - v_0\right) \cdot c_{i,0} - v_1\right) \cdot c_{i,1} - \ldots - v_{i-1}\right) \cdot c_{i,i-1} \bmod m_i$$

### Phase 2: Reconstruction

$$x = v_0 + v_1 \cdot m_0 + v_2 \cdot m_0 \cdot m_1 + \ldots + v_{n-1} \cdot \prod_{j=0}^{n-2} m_j$$

## 🔐 Garner Constants Precomputation

Before applying MRC, we must compute the **Garner constants** $c_{i,j}$:

$$c_{i,j} = m_j^{-1} \bmod m_i \quad \text{for all } j < i$$

**Example with moduli** $\{2, 3, 5\}$:

- $c_{1,0} = 2^{-1} \bmod 3 = 2$ (since $2 \times 2 \equiv 1 \pmod{3}$)
- $c_{2,0} = 2^{-1} \bmod 5 = 3$ (since $2 \times 3 \equiv 1 \pmod{5}$)
- $c_{2,1} = 3^{-1} \bmod 5 = 2$ (since $3 \times 2 \equiv 1 \pmod{5}$)

These constants are computed once and reused for all conversions.

</div>

---

<div style="background: #e7f3ff; padding: 20px; border-left: 5px solid #2196F3; border-radius: 8px; margin: 20px 0;">

## 📊 Step-by-Step Example

**Given:** Moduli $\{2, 3, 5\}$, RNS representation $[1, 2, 3]$

### Step 1: Compute v₀
$$v_0 = r_0 = 1$$

### Step 2: Compute v₁
$$\begin{align}
v_1 &= (r_1 - v_0) \cdot c_{1,0} \bmod m_1 \\
&= (2 - 1) \cdot 2 \bmod 3 \\
&= 2
\end{align}$$

### Step 3: Compute v₂
$$\begin{align}
\text{temp} &= (r_2 - v_0) \cdot c_{2,0} \bmod m_2 = (3 - 1) \cdot 3 \bmod 5 = 1 \\
v_2 &= (\text{temp} - v_1) \cdot c_{2,1} \bmod m_2 = (1 - 2) \cdot 2 \bmod 5 = 3
\end{align}$$

### Step 4: Reconstruct
$$\begin{align}
x &= v_0 + v_1 \cdot m_0 + v_2 \cdot m_0 \cdot m_1 \\
&= 1 + 2 \cdot 2 + 3 \cdot 2 \cdot 3 \\
&= 1 + 4 + 18 = 23
\end{align}$$

<div style="background: linear-gradient(135deg, #ffeaa7 0%, #fdcb6e 100%); padding: 20px; border-radius: 10px; margin: 20px 0;">

## ⚖️ Advantages of MRC over CRT

| Aspect | CRT | MRC (Garner) |
|--------|-----|--------------|
| **Computation** | Parallel | Sequential |
| **Intermediate Values** | Large (up to M²) | Small (< M) |
| **Hardware** | More complex | Simpler |
| **Precomputation** | M<sub>i</sub>, y<sub>i</sub> | c<sub>i,j</sub> |

**Key Advantage:** MRC avoids large intermediate values, making it more suitable for hardware implementation.

</div>

---

<div style="background: #f8d7da; padding: 20px; border-left: 5px solid #dc3545; border-radius: 8px; margin: 20px 0;">

## ⚠️ Important Notes

- **Uniqueness:** Like CRT, MRC guarantees unique reconstruction for $x \in [0, M-1]$
- **Signed Numbers:** Apply the same mapping as CRT: if $x > \lfloor(M-1)/2\rfloor$, then signed value is $x - M$
- **Complexity:** MRC requires $O(n²)$ modular multiplications vs CRT's $O(n)$ but with smaller operands

</div>

```python
# ============================================================
# Part 3: Inverse Conversion using Mixed Radix Conversion (MRC)
# Garner's Algorithm Implementation
# ============================================================

def compute_garner_constants(moduli: List[int]) -> List[List[int]]:
    """
    Precompute the constants needed for Garner's algorithm.

    For MRC, we need to compute:
    c[i][j] = (m_j)^(-1) mod m_i for all j < i

    These constants are used to convert RNS to mixed-radix representation.

    Args:
        moduli: List of pairwise coprime moduli [m1, m2, ..., mn]

    Returns:
        2D list of constants where c[i][j] = modular inverse of m_j mod m_i
    """
    n = len(moduli)
    c = [[0] * n for _ in range(n)]

    for i in range(1, n):
        for j in range(i):
            c[i][j] = modular_inverse(moduli[j], moduli[i])

    return c

def mrc_inverse_unsigned(residues: List[int], moduli: List[int],
                          garner_constants: List[List[int]] = None) -> int:
    """
    Convert RNS representation back to unsigned integer using
    Mixed Radix Conversion (Garner's Algorithm).

    Garner's Algorithm:
    1. Compute mixed-radix digits v[i] iteratively
    2. v[0] = r[0]
    3. v[i] = ((...((r[i] - v[0]) * c[i][0] - v[1]) * c[i][1] - ...) * c[i][i-1]) mod m[i]
    4. Final result: x = v[0] + v[1]*m[0] + v[2]*m[0]*m[1] + ...

    Args:
        residues: List of residues [r1, r2, ..., rn]
        moduli: List of pairwise coprime moduli [m1, m2, ..., mn]
        garner_constants: Precomputed constants (optional, will compute if not provided)

    Returns:
        Unsigned integer x in range [0, M-1]
    """
    n = len(moduli)

    # Compute Garner constants if not provided
    if garner_constants is None:
        garner_constants = compute_garner_constants(moduli)

    # Step 1: Compute mixed-radix digits v[i]
    v = [0] * n
    v[0] = residues[0] % moduli[0]

    for i in range(1, n):
        # Start with residue[i]
        temp = residues[i]

        # Subtract and multiply by inverse iteratively
        for j in range(i):
            # temp = (temp - v[j]) * c[i][j] mod m[i]
            temp = (temp - v[j]) % moduli[i]
            temp = (temp * garner_constants[i][j]) % moduli[i]

        v[i] = temp

    # Step 2: Reconstruct the number from mixed-radix representation
    # x = v[0] + v[1]*m[0] + v[2]*m[0]*m[1] + v[3]*m[0]*m[1]*m[2] + ...
    result = 0
    multiplier = 1

    for i in range(n):
        result += v[i] * multiplier
        multiplier *= moduli[i]

    return result

def mrc_inverse_signed(residues: List[int], moduli: List[int],
                        garner_constants: List[List[int]] = None) -> int:
    """
    Convert RNS representation back to signed integer using MRC.

    Similar to CRT signed conversion:
    - First get unsigned result
    - If result >= M/2, subtract M to get negative value

    Args:
        residues: List of residues [r1, r2, ..., rn]
        moduli: List of pairwise coprime moduli [m1, m2, ..., mn]
        garner_constants: Precomputed constants (optional)

    Returns:
        Signed integer x in range [-M//2, (M-1)//2]
    """
    M = compute_dynamic_range(moduli)

    # Get unsigned result using MRC
    unsigned_result = mrc_inverse_unsigned(residues, moduli, garner_constants)

    # Convert to signed representation
    if unsigned_result > (M - 1) // 2:
        return unsigned_result - M
    else:
        return unsigned_result

def demonstrate_garner_step_by_step(residues: List[int], moduli: List[int]):
    """
    Demonstrate Garner's algorithm step by step for educational purposes.
    """
    n = len(moduli)
    garner_constants = compute_garner_constants(moduli)

    print(f"Input residues: {residues}")
    print(f"Moduli: {moduli}")
    print()

    # Show Garner constants
    print("Garner Constants c[i][j] = m_j^(-1) mod m_i:")
    for i in range(1, n):
        for j in range(i):
            print(f"  c[{i}][{j}] = {moduli[j]}^(-1) mod {moduli[i]} = {garner_constants[i][j]}")
    print()

    # Compute mixed-radix digits
    v = [0] * n
    v[0] = residues[0] % moduli[0]
    print(f"Mixed-radix digit v[0] = r[0] mod m[0] = {residues[0]} mod {moduli[0]} = {v[0]}")

    for i in range(1, n):
        temp = residues[i]
        print(f"\nComputing v[{i}]:")
        print(f"  Start: temp = r[{i}] = {residues[i]}")

        for j in range(i):
            old_temp = temp
            temp = (temp - v[j]) % moduli[i]
            temp = (temp * garner_constants[i][j]) % moduli[i]
            print(f"  Step {j+1}: temp = ({old_temp} - v[{j}]) * c[{i}][{j}] mod m[{i}]")
            print(f"         = ({old_temp} - {v[j]}) * {garner_constants[i][j]} mod {moduli[i]} = {temp}")

        v[i] = temp
        print(f"  Result: v[{i}] = {v[i]}")

    # Reconstruct number
    print(f"\nMixed-radix digits: v = {v}")
    print(f"\nReconstruction: x = v[0] + v[1]*m[0] + v[2]*m[0]*m[1] + ...")

    result = 0
    multiplier = 1
    terms = []

    for i in range(n):
        term = v[i] * multiplier
        terms.append(f"{v[i]}*{multiplier}")
        result += term
        multiplier *= moduli[i]

    print(f"x = {' + '.join(terms)} = {result}")

    return result


# ==================== Testing MRC Inverse Conversion ====================

print("=" * 70)
print("Part 3: Inverse Conversion using Mixed Radix Conversion (MRC)")
print("         Garner's Algorithm Implementation")
print("=" * 70)

moduli = [2, 3, 5]
M = compute_dynamic_range(moduli)

print(f"\nModuli set: {moduli}")
print(f"Dynamic range M = {M}")

# Show Garner constants
print("\n" + "-" * 70)
print("GARNER CONSTANTS")
print("-" * 70)
garner_constants = compute_garner_constants(moduli)
print("c[i][j] = (m_j)^(-1) mod m_i for j < i:\n")
for i in range(1, len(moduli)):
    for j in range(i):
        print(f"c[{i}][{j}] = {moduli[j]}^(-1) mod {moduli[i]} = {garner_constants[i][j]}")
        print(f"  Verification: {moduli[j]} × {garner_constants[i][j]} mod {moduli[i]} = {(moduli[j] * garner_constants[i][j]) % moduli[i]}")

# Step-by-step demonstration for one example
print("\n" + "-" * 70)
print("STEP-BY-STEP DEMONSTRATION (Example: x = 23)")
print("-" * 70)
example_x = 23
example_rns = forward_conversion_unsigned(example_x, moduli)
print(f"\nOriginal number: {example_x}")
print(f"RNS representation: {example_rns}\n")
demonstrate_garner_step_by_step(example_rns, moduli)

# Unsigned numbers - MRC verification
print("\n" + "-" * 70)
print("UNSIGNED NUMBERS - MRC INVERSE VERIFICATION")
print("-" * 70)
unsigned_start, unsigned_end = get_unsigned_range(moduli)
print(f"Valid range: [{unsigned_start}, {unsigned_end}]\n")

print(f"{'Original':<10} {'RNS':<20} {'MRC Result':<12} {'CRT Result':<12} {'Match'}")
print("-" * 70)

all_match_unsigned = True
for x in range(unsigned_start, unsigned_end + 1):
    rns = forward_conversion_unsigned(x, moduli)
    mrc_result = mrc_inverse_unsigned(rns, moduli, garner_constants)
    crt_result = crt_inverse_unsigned(rns, moduli)
    match = (x == mrc_result == crt_result)
    all_match_unsigned = all_match_unsigned and match
    status = "✓" if match else "✗"
    print(f"{x:<10} {str(rns):<20} {mrc_result:<12} {crt_result:<12} {status}")

print(f"\nAll unsigned MRC conversions correct: {'PASSED ✓' if all_match_unsigned else 'FAILED ✗'}")

# Signed numbers - MRC verification
print("\n" + "-" * 70)
print("SIGNED NUMBERS - MRC INVERSE VERIFICATION")
print("-" * 70)
signed_start, signed_end = get_signed_range(moduli)
print(f"Valid range: [{signed_start}, {signed_end}]\n")

print(f"{'Original':<10} {'RNS':<20} {'MRC Result':<12} {'CRT Result':<12} {'Match'}")
print("-" * 70)

all_match_signed = True
for x in range(signed_start, signed_end + 1):
    rns = forward_conversion_signed(x, moduli)
    mrc_result = mrc_inverse_signed(rns, moduli, garner_constants)
    crt_result = crt_inverse_signed(rns, moduli)
    match = (x == mrc_result == crt_result)
    all_match_signed = all_match_signed and match
    status = "✓" if match else "✗"
    print(f"{x:<10} {str(rns):<20} {mrc_result:<12} {crt_result:<12} {status}")

print(f"\nAll signed MRC conversions correct: {'PASSED ✓' if all_match_signed else 'FAILED ✗'}")

# Summary
print("\n" + "=" * 70)
print("PART 3 SUMMARY")
print("=" * 70)
print("Garner's Algorithm for Mixed Radix Conversion:")
print()
print("1. Compute Garner constants: c[i][j] = m_j^(-1) mod m_i")
print("2. Compute mixed-radix digits v[i] iteratively:")
print("   v[0] = r[0]")
print("   v[i] = ((...((r[i] - v[0]) × c[i][0] - v[1]) × c[i][1] ...) × c[i][i-1]) mod m[i]")
print("3. Reconstruct: x = v[0] + v[1]×m[0] + v[2]×m[0]×m[1] + ...")
print()
print(f"Unsigned conversion verification: {'PASSED ✓' if all_match_unsigned else 'FAILED ✗'}")
print(f"Signed conversion verification: {'PASSED ✓' if all_match_signed else 'FAILED ✗'}")
print(f"MRC results match CRT results: {'PASSED ✓' if all_match_unsigned and all_match_signed else 'FAILED ✗'}")
```

## 🔑 Key Algorithm Components

<div style="background: white; padding: 20px; border-radius: 10px; margin: 15px 0;">

### Garner's Algorithm Formula

**Step 1: Initialize**
<div style="text-align: center; font-size: 18px; padding: 15px; background: #fce4ec; border-radius: 8px; margin: 10px 0;">
<strong>v[0] = r[0] mod m[0]</strong>
</div>

**Step 2: Iterative Computation (for i = 1 to n-1)**
<div style="text-align: center; font-size: 18px; padding: 15px; background: #fce4ec; border-radius: 8px; margin: 10px 0;">
<strong>v[i] = ((...((r[i] - v[0]) × c[i][0] - v[1]) × c[i][1] - ...) × c[i][i-1]) mod m[i]</strong>
</div>

**Step 3: Reconstruction**
<div style="text-align: center; font-size: 18px; padding: 15px; background: #fce4ec; border-radius: 8px; margin: 10px 0;">
<strong>x = v[0] + v[1]×m[0] + v[2]×m[0]×m[1] + ...</strong>
</div>

</div>

<div style="background: white; padding: 20px; border-radius: 10px; margin: 15px 0;">

### Example Trace (x = 23)

**Given:** RNS = [1, 2, 3], Moduli = [2, 3, 5]
v[0] = 1

v[1]: temp = 2
temp = (2 - 1) × 2 mod 3 = 2
v[1] = 2

v[2]: temp = 3
temp = (3 - 1) × 3 mod 5 = 1
temp = (1 - 2) × 2 mod 5 = 3
v[2] = 3

Reconstruction:
x = 1 + 2×2 + 3×6 = 1 + 4 + 18 = 23 ✓

# **Part 4: Arithmetic Operations in RNS Domain**

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 30px; border-radius: 15px; margin: 20px 0; box-shadow: 0 10px 25px rgba(0,0,0,0.2);">

        Addition, Subtraction, and Multiplication in Residue Number System
    </h2>
</div>

<div style="background: linear-gradient(to right, #f8f9fa, #e9ecef); padding: 25px; border-left: 5px solid #667eea; border-radius: 10px; margin: 20px 0;">

## 📚 Theoretical Foundation

One of the most powerful features of RNS is the ability to perform arithmetic operations **independently** on each residue channel. This property enables **parallel computation** without carry propagation between channels, making RNS particularly attractive for high-speed arithmetic applications.

### 🎯 Core Principle

For any arithmetic operation **⊕** (addition, subtraction, or multiplication) on two numbers **a** and **b**:

<div style="text-align: center; padding: 20px; background: white; border-radius: 10px; margin: 15px 0; box-shadow: 0 2px 8px rgba(0,0,0,0.1);">
<strong style="font-size: 1.3em; color: #667eea;">
RNS(a ⊕ b) = [RNS(a) ⊕ RNS(b)] mod mᵢ
</strong>
</div>

This means each residue can be computed **independently** in parallel!

</div>

---

<div style="background: #d1fae5; padding: 25px; border-left: 5px solid #10b981; border-radius: 10px; margin: 20px 0;">

## ➕ Addition in RNS Domain

### Mathematical Formula

For two numbers **a** and **b** with RNS representations:
- **RNS(a) = [a₁, a₂, ..., aₙ]**
- **RNS(b) = [b₁, b₂, ..., bₙ]**

The sum in RNS is computed as:

<div style="text-align: center; padding: 15px; background: white; border-radius: 8px; margin: 15px 0;">
<strong>RNS(a + b) = [(a₁ + b₁) mod m₁, (a₂ + b₂) mod m₂, ..., (aₙ + bₙ) mod mₙ]</strong>
</div>

### Key Properties

<ul style="line-height: 2; font-size: 16px;">
<li><strong>✓ No Carry Propagation:</strong> Each channel operates independently</li>
<li><strong>✓ Parallel Execution:</strong> All n additions can occur simultaneously</li>
<li><strong>✓ Modular Arithmetic:</strong> Results automatically stay within valid range</li>
<li><strong>✓ Speed:</strong> O(1) complexity with parallel hardware</li>
</ul>

</div>

---

<div style="background: #fee2e2; padding: 25px; border-left: 5px solid #ef4444; border-radius: 10px; margin: 20px 0;">

## ➖ Subtraction in RNS Domain

### Mathematical Formula

For subtraction **a - b**:

<div style="text-align: center; padding: 15px; background: white; border-radius: 8px; margin: 15px 0;">
<strong>RNS(a - b) = [(a₁ - b₁) mod m₁, (a₂ - b₂) mod m₂, ..., (aₙ - bₙ) mod mₙ]</strong>
</div>

### Handling Negative Residues

When **(aᵢ - bᵢ) < 0**, we use modular arithmetic:

<div style="background: white; padding: 15px; border-radius: 8px; margin: 15px 0;">
<strong>If (aᵢ - bᵢ) < 0:</strong> Add mᵢ to make it positive<br>
<strong>Example:</strong> (3 - 7) mod 5 = -4 mod 5 = 1
</div>

### Important Note

<div style="background: #fef3c7; padding: 15px; border-radius: 8px; margin: 15px 0; border-left: 4px solid #f59e0b;">
⚠️ <strong>Overflow Detection:</strong> RNS cannot easily detect when results exceed the dynamic range M. Care must be taken to ensure operands are chosen such that results remain valid.
</div>

</div>

## ✖️ Multiplication in RNS Domain

### Mathematical Formula

For multiplication **a × b**:

<div style="text-align: center; padding: 15px; background: white; border-radius: 8px; margin: 15px 0;">
<strong>RNS(a × b) = [(a₁ × b₁) mod m₁, (a₂ × b₂) mod m₂, ..., (aₙ × bₙ) mod mₙ]</strong>
</div>

### Advantages

<ul style="line-height: 2; font-size: 16px;">
<li><strong>✓ No Partial Products:</strong> Direct modular multiplication</li>
<li><strong>✓ Smaller Operands:</strong> Each multiplication is (aᵢ × bᵢ) where aᵢ, bᵢ < mᵢ</li>
<li><strong>✓ Parallel Execution:</strong> All channels compute independently</li>
<li><strong>✓ Hardware Friendly:</strong> Small multipliers can be used</li>
</ul>

### Constraint

<div style="background: #fef3c7; padding: 15px; border-radius: 8px; margin: 15px 0; border-left: 4px solid #f59e0b;">
⚠️ <strong>Range Limitation:</strong> For multiplication, ensure a × b < M to avoid overflow. This typically means choosing operands from a restricted range.
</div>

## 🔬 Practical Example

**Numbers:** a = 7, b = 5

### Forward Conversion
- **RNS(7) = [7 mod 3, 7 mod 4, 7 mod 5] = [1, 3, 2]**
- **RNS(5) = [5 mod 3, 5 mod 4, 5 mod 5] = [2, 1, 0]**

### Addition: 7 + 5 = 12
<div style="background: white; padding: 15px; border-radius: 8px; margin: 10px 0; font-family: monospace;">

Inverse: 12 ✓

           = [2, 2, 2]<br>
Inverse: 2 ✓
</div>

### Multiplication: 7 × 5 = 35
<div style="background: white; padding: 15px; border-radius: 8px; margin: 10px 0; font-family: monospace;">

           = [2, 3, 0]<br>
Inverse: 35 ✓

```python
# ============================================================
# Section 4: Arithmetic Operations in RNS Domain
# ============================================================

import random

def rns_add(rns_a: List[int], rns_b: List[int], moduli: List[int]) -> List[int]:
    """
    Addition in RNS domain.

    In RNS, addition is performed component-wise:
    (a + b) mod mi for each modulus mi

    Args:
        rns_a: First operand in RNS representation
        rns_b: Second operand in RNS representation
        moduli: List of pairwise coprime moduli

    Returns:
        Sum in RNS representation
    """
    result = []
    for i in range(len(moduli)):
        result.append((rns_a[i] + rns_b[i]) % moduli[i])
    return result


def rns_subtract(rns_a: List[int], rns_b: List[int], moduli: List[int]) -> List[int]:
    """
    Subtraction in RNS domain.

    In RNS, subtraction is performed component-wise:
    (a - b) mod mi for each modulus mi

    Args:
        rns_a: First operand in RNS representation (minuend)
        rns_b: Second operand in RNS representation (subtrahend)
        moduli: List of pairwise coprime moduli

    Returns:
        Difference in RNS representation
    """
    result = []
    for i in range(len(moduli)):
        # Ensure non-negative result by adding modulus before taking mod
        result.append((rns_a[i] - rns_b[i] + moduli[i]) % moduli[i])
    return result


def rns_multiply(rns_a: List[int], rns_b: List[int], moduli: List[int]) -> List[int]:
    """
    Multiplication in RNS domain.

    In RNS, multiplication is performed component-wise:
    (a * b) mod mi for each modulus mi

    Args:
        rns_a: First operand in RNS representation
        rns_b: Second operand in RNS representation
        moduli: List of pairwise coprime moduli

    Returns:
        Product in RNS representation
    """
    result = []
    for i in range(len(moduli)):
        result.append((rns_a[i] * rns_b[i]) % moduli[i])
    return result


def generate_test_pairs_addition(moduli: List[int], count: int = 5) -> List[Tuple[int, int]]:
    """
    Generate random test pairs for addition such that result doesn't overflow.

    For unsigned: 0 <= a + b < M
    """
    M = compute_dynamic_range(moduli)
    pairs = []

    random.seed(42)  # For reproducibility

    for _ in range(count):
        # Generate a in range [0, M-1]
        a = random.randint(0, M // 2 - 1)
        # Generate b such that a + b < M
        b = random.randint(0, M - a - 1)
        pairs.append((a, b))

    return pairs


def generate_test_pairs_subtraction(moduli: List[int], count: int = 5) -> List[Tuple[int, int]]:
    """
    Generate random test pairs for subtraction such that result doesn't underflow.

    For unsigned: a >= b so that a - b >= 0
    """
    M = compute_dynamic_range(moduli)
    pairs = []

    random.seed(43)  # For reproducibility

    for _ in range(count):
        # Generate a in range [1, M-1]
        a = random.randint(1, M - 1)
        # Generate b such that b <= a
        b = random.randint(0, a)
        pairs.append((a, b))

    return pairs


def generate_test_pairs_multiplication(moduli: List[int], count: int = 5) -> List[Tuple[int, int]]:
    """
    Generate random test pairs for multiplication such that result doesn't overflow.

    For unsigned: a * b < M
    """
    M = compute_dynamic_range(moduli)
    pairs = []

    random.seed(44)  # For reproducibility

    for _ in range(count):
        # Generate a in range [1, sqrt(M)]
        max_a = int(M ** 0.5)
        a = random.randint(1, max_a)
        # Generate b such that a * b < M
        max_b = (M - 1) // a
        b = random.randint(1, max_b)
        pairs.append((a, b))

    return pairs


# ============================================================
# Testing with moduli {3, 4, 5}
# ============================================================

moduli = [3, 4, 5]
M = compute_dynamic_range(moduli)

print("=" * 80)
print("Section 4: Arithmetic Operations in RNS Domain")
print("=" * 80)
print(f"Moduli set: {moduli}")
print(f"Dynamic range M = {' × '.join(map(str, moduli))} = {M}")
print()

# -----------------------------
# Test 1: Addition
# -----------------------------
print("-" * 80)
print("Test 1: ADDITION in RNS")
print("-" * 80)
print(f"{'A':<6} {'B':<6} {'A+B':<6} {'RNS(A)':<15} {'RNS(B)':<15} {'RNS(A)+RNS(B)':<15} {'Inverse':<8} {'Match'}")
print("-" * 80)

add_pairs = generate_test_pairs_addition(moduli, 5)
all_add_correct = True

for a, b in add_pairs:
    expected_sum = a + b

    # Convert to RNS
    rns_a = forward_conversion_unsigned(a, moduli)
    rns_b = forward_conversion_unsigned(b, moduli)

    # Perform addition in RNS
    rns_sum = rns_add(rns_a, rns_b, moduli)

    # Convert back to verify
    recovered_sum = crt_inverse_unsigned(rns_sum, moduli)

    correct = (expected_sum == recovered_sum)
    if not correct:
        all_add_correct = False

    status = "✓" if correct else "✗"
    print(f"{a:<6} {b:<6} {expected_sum:<6} {str(rns_a):<15} {str(rns_b):<15} {str(rns_sum):<15} {recovered_sum:<8} {status}")

print("-" * 80)
print(f"All addition tests passed: {'PASSED ✓' if all_add_correct else 'FAILED ✗'}")
print()

# -----------------------------
# Test 2: Subtraction
# -----------------------------
print("-" * 80)
print("Test 2: SUBTRACTION in RNS")
print("-" * 80)
print(f"{'A':<6} {'B':<6} {'A-B':<6} {'RNS(A)':<15} {'RNS(B)':<15} {'RNS(A)-RNS(B)':<15} {'Inverse':<8} {'Match'}")
print("-" * 80)

sub_pairs = generate_test_pairs_subtraction(moduli, 5)
all_sub_correct = True

for a, b in sub_pairs:
    expected_diff = a - b

    # Convert to RNS
    rns_a = forward_conversion_unsigned(a, moduli)
    rns_b = forward_conversion_unsigned(b, moduli)

    # Perform subtraction in RNS
    rns_diff = rns_subtract(rns_a, rns_b, moduli)

    # Convert back to verify
    recovered_diff = crt_inverse_unsigned(rns_diff, moduli)

    correct = (expected_diff == recovered_diff)
    if not correct:
        all_sub_correct = False

    status = "✓" if correct else "✗"
    print(f"{a:<6} {b:<6} {expected_diff:<6} {str(rns_a):<15} {str(rns_b):<15} {str(rns_diff):<15} {recovered_diff:<8} {status}")

print("-" * 80)
print(f"All subtraction tests passed: {'PASSED ✓' if all_sub_correct else 'FAILED ✗'}")
print()

# -----------------------------
# Test 3: Multiplication
# -----------------------------
print("-" * 80)
print("Test 3: MULTIPLICATION in RNS")
print("-" * 80)
print(f"{'A':<6} {'B':<6} {'A×B':<6} {'RNS(A)':<15} {'RNS(B)':<15} {'RNS(A)×RNS(B)':<15} {'Inverse':<8} {'Match'}")
print("-" * 80)

mul_pairs = generate_test_pairs_multiplication(moduli, 5)
all_mul_correct = True

for a, b in mul_pairs:
    expected_prod = a * b

    # Convert to RNS
    rns_a = forward_conversion_unsigned(a, moduli)
    rns_b = forward_conversion_unsigned(b, moduli)

    # Perform multiplication in RNS
    rns_prod = rns_multiply(rns_a, rns_b, moduli)

    # Convert back to verify
    recovered_prod = crt_inverse_unsigned(rns_prod, moduli)

    correct = (expected_prod == recovered_prod)
    if not correct:
        all_mul_correct = False

    status = "✓" if correct else "✗"
    print(f"{a:<6} {b:<6} {expected_prod:<6} {str(rns_a):<15} {str(rns_b):<15} {str(rns_prod):<15} {recovered_prod:<8} {status}")

print("-" * 80)
print(f"All multiplication tests passed: {'PASSED ✓' if all_mul_correct else 'FAILED ✗'}")
print()

# -----------------------------
# Summary
# -----------------------------
print("=" * 80)
print("Section 4 Summary")
print("=" * 80)
print(f"Moduli: {moduli}")
print(f"Dynamic Range (M): {M}")
print()
print("RNS Arithmetic Properties:")
print("  • Addition:       (a + b) mod mi  for each modulus mi")
print("  • Subtraction:    (a - b) mod mi  for each modulus mi")
print("  • Multiplication: (a × b) mod mi  for each modulus mi")
print()
print("Test Results:")
print(f"  • Addition (5 random pairs):       {'PASSED ✓' if all_add_correct else 'FAILED ✗'}")
print(f"  • Subtraction (5 random pairs):    {'PASSED ✓' if all_sub_correct else 'FAILED ✗'}")
print(f"  • Multiplication (5 random pairs): {'PASSED ✓' if all_mul_correct else 'FAILED ✗'}")
print()
print("Key Advantage of RNS:")
print("  All operations are performed independently on each residue,")
print("  enabling parallel computation without carry propagation.")
print("=" * 80)
```

## 📚 Introduction to CNN in RNS

**Convolutional Neural Networks (CNNs)** are the backbone of modern computer vision and image processing applications. However, their computational intensity makes them challenging to deploy on resource-constrained embedded systems. **RNS** offers a promising solution by enabling:

- **Parallel arithmetic operations** without carry propagation
- **Reduced bit-width computations** through modular channels
- **Energy-efficient hardware implementations**
- **Natural overflow handling** through modular arithmetic

### 🎯 Motivation

In this part, we explore how RNS can accelerate CNN inference by implementing two fundamental operations:

1. **Convolution:** The core operation in CNNs for feature extraction
2. **ReLU Activation:** The most common non-linear activation function

By performing these operations entirely in the RNS domain, we can leverage parallel processing and avoid expensive conversions between domains during computation.

- Perform MAC operations per modulus channel
- Results are automatically mod M

**2. ReLU in RNS:**
- Challenge: Sign detection requires conversion
- Solution: Convert to real → apply ReLU → convert back

**3. Verification:**
- Compare: `(real_output mod M)` vs `rns_output`

## Implementing Convolution and ReLU Operations in RNS Domain

```python
# ============================================================
# Part 5: RNS Application in CNN Neural Network
# ============================================================

import numpy as np
import random

# Set random seed for reproducibility
np.random.seed(42)
random.seed(42)

# ============================================================
# RNS Helper Functions (Using functions from previous sections)
# ============================================================

def rns_forward(x: int, moduli: List[int]) -> List[int]:
    """
    Convert integer to RNS representation.
    Handles negative numbers by adding M first.
    """
    M = compute_dynamic_range(moduli)
    if x < 0:
        x = x + M
    return [x % m for m in moduli]

def rns_inverse(residues: List[int], moduli: List[int]) -> int:
    """
    Convert RNS representation back to integer using CRT.
    Returns signed result.
    """
    return crt_inverse_signed(residues, moduli)

def rns_add(a_rns: List[int], b_rns: List[int], moduli: List[int]) -> List[int]:
    """
    Add two numbers in RNS domain.
    """
    return [(a + b) % m for a, b, m in zip(a_rns, b_rns, moduli)]

def rns_multiply(a_rns: List[int], b_rns: List[int], moduli: List[int]) -> List[int]:
    """
    Multiply two numbers in RNS domain.
    """
    return [(a * b) % m for a, b, m in zip(a_rns, b_rns, moduli)]

# ============================================================
# CNN Components in Real Domain
# ============================================================

def conv2d_real(input_tensor: np.ndarray, kernel: np.ndarray) -> np.ndarray:
    """
    Perform 2D convolution in real domain.

    Args:
        input_tensor: Input with shape (H, W, C)
        kernel: Convolution kernel with shape (kH, kW, C)

    Returns:
        Output feature map with shape (out_H, out_W)
    """
    H, W, C = input_tensor.shape
    kH, kW, kC = kernel.shape

    assert C == kC, "Input channels must match kernel channels"

    # Output dimensions (no padding, stride=1)
    out_H = H - kH + 1
    out_W = W - kW + 1

    output = np.zeros((out_H, out_W), dtype=np.int64)

    for i in range(out_H):
        for j in range(out_W):
            # Extract patch
            patch = input_tensor[i:i+kH, j:j+kW, :]
            # Compute convolution (element-wise multiply and sum)
            output[i, j] = np.sum(patch * kernel)

    return output

def relu_real(x: np.ndarray) -> np.ndarray:
    """
    Apply ReLU activation in real domain.
    """
    return np.maximum(0, x)

# ============================================================
# CNN Components in RNS Domain
# ============================================================

def conv2d_rns(input_rns: np.ndarray, kernel_rns: np.ndarray,
               moduli: List[int]) -> np.ndarray:
    """
    Perform 2D convolution in RNS domain.

    Args:
        input_rns: Input with shape (H, W, C, num_moduli)
        kernel_rns: Kernel with shape (kH, kW, C, num_moduli)
        moduli: List of moduli

    Returns:
        Output feature map with shape (out_H, out_W, num_moduli)
    """
    H, W, C, num_moduli = input_rns.shape
    kH, kW, kC, _ = kernel_rns.shape

    out_H = H - kH + 1
    out_W = W - kW + 1

    output_rns = np.zeros((out_H, out_W, num_moduli), dtype=np.int64)

    for i in range(out_H):
        for j in range(out_W):
            # Initialize accumulator in RNS (represents 0)
            acc_rns = [0] * num_moduli

            # Convolution in RNS domain
            for ki in range(kH):
                for kj in range(kW):
                    for c in range(C):
                        # Get RNS representations
                        input_val_rns = list(input_rns[i+ki, j+kj, c, :])
                        kernel_val_rns = list(kernel_rns[ki, kj, c, :])

                        # Multiply in RNS
                        product_rns = rns_multiply(input_val_rns, kernel_val_rns, moduli)

                        # Add to accumulator in RNS
                        acc_rns = rns_add(acc_rns, product_rns, moduli)

            output_rns[i, j, :] = acc_rns

    return output_rns

def relu_rns(x_rns: np.ndarray, moduli: List[int]) -> np.ndarray:
    """
    Apply ReLU activation in RNS domain.

    For ReLU, we need to determine if the number is negative.
    In signed RNS, numbers >= M/2 represent negative values.

    Args:
        x_rns: Input with shape (..., num_moduli)
        moduli: List of moduli

    Returns:
        Output with ReLU applied
    """
    M = compute_dynamic_range(moduli)
    output_shape = x_rns.shape
    output_rns = np.zeros_like(x_rns)

    # Flatten all dimensions except last (moduli dimension)
    flat_shape = (-1, len(moduli))
    x_flat = x_rns.reshape(flat_shape)
    output_flat = output_rns.reshape(flat_shape)

    for idx in range(x_flat.shape[0]):
        residues = list(x_flat[idx, :])
        # Convert to real value to check sign
        real_val = crt_inverse_unsigned(residues, moduli)

        # Check if negative (>= M/2 in unsigned representation)
        if real_val > (M - 1) // 2:
            # Negative number, ReLU outputs 0
            output_flat[idx, :] = [0] * len(moduli)
        else:
            # Positive number, keep as is
            output_flat[idx, :] = residues

    return output_flat.reshape(output_shape)

# ============================================================
# Conversion Functions for Tensors
# ============================================================

def tensor_to_rns(tensor: np.ndarray, moduli: List[int]) -> np.ndarray:
    """
    Convert a tensor of integers to RNS representation.

    Args:
        tensor: Input tensor of integers
        moduli: List of moduli

    Returns:
        Tensor with additional dimension for RNS residues
    """
    output_shape = tensor.shape + (len(moduli),)
    output = np.zeros(output_shape, dtype=np.int64)

    for idx in np.ndindex(tensor.shape):
        val = int(tensor[idx])
        rns_val = rns_forward(val, moduli)
        output[idx] = rns_val

    return output

def tensor_from_rns(tensor_rns: np.ndarray, moduli: List[int],
                    signed: bool = True) -> np.ndarray:
    """
    Convert a tensor from RNS representation back to integers.

    Args:
        tensor_rns: Input tensor with last dimension being RNS residues
        moduli: List of moduli
        signed: If True, return signed integers

    Returns:
        Tensor of integers
    """
    output_shape = tensor_rns.shape[:-1]
    output = np.zeros(output_shape, dtype=np.int64)

    for idx in np.ndindex(output_shape):
        residues = list(tensor_rns[idx])
        if signed:
            output[idx] = crt_inverse_signed(residues, moduli)
        else:
            output[idx] = crt_inverse_unsigned(residues, moduli)

    return output

# ============================================================
# Main CNN Experiment
# ============================================================

print("=" * 70)
print("Part 5: RNS Application in CNN Neural Network")
print("=" * 70)

# Define moduli set
moduli = [7, 8, 9]
M = compute_dynamic_range(moduli)  # M = 7 * 8 * 9 = 504

print(f"\nModuli set: {moduli}")
print(f"Dynamic range M = {M}")
print(f"Weight/Input range: [-256, 255]")

# Verify dynamic range is sufficient
print(f"\nNote: M = {M}, which allows representing values in range [-{M//2}, {M//2 - 1}]")
print(f"      = [-252, 251] for signed representation")

# ============================================================
# Generate Random Input and Weights
# ============================================================

print("\n" + "-" * 70)
print("GENERATING RANDOM INPUT AND WEIGHTS")
print("-" * 70)

# Input: 8x8x3
input_shape = (8, 8, 3)
input_tensor = np.random.randint(-256, 256, size=input_shape, dtype=np.int64)

# Kernel: 5x5x3
kernel_shape = (5, 5, 3)
kernel = np.random.randint(-256, 256, size=kernel_shape, dtype=np.int64)

print(f"Input shape: {input_tensor.shape}")
print(f"Kernel shape: {kernel.shape}")
print(f"Expected output shape: (4, 4)")

print(f"\nInput tensor sample (first channel, first 4x4):")
print(input_tensor[:4, :4, 0])

print(f"\nKernel sample (first channel, first 3x3):")
print(kernel[:3, :3, 0])

# ============================================================
# Convolution in Real Domain
# ============================================================

print("\n" + "-" * 70)
print("CONVOLUTION IN REAL DOMAIN")
print("-" * 70)

# Perform convolution
conv_output_real = conv2d_real(input_tensor, kernel)
print(f"Convolution output (before ReLU):")
print(conv_output_real)

# Apply ReLU
relu_output_real = relu_real(conv_output_real)
print(f"\nReLU output:")
print(relu_output_real)

# Convert to modulo M for comparison with RNS
conv_output_real_mod = conv_output_real % M
# Handle negative values for modulo
conv_output_real_mod = np.where(conv_output_real >= 0,
                                 conv_output_real % M,
                                 (conv_output_real % M + M) % M)

print(f"\nConvolution output (mod {M}, for RNS comparison):")
print(conv_output_real_mod)

# ============================================================
# Convolution in RNS Domain
# ============================================================

print("\n" + "-" * 70)
print("CONVOLUTION IN RNS DOMAIN")
print("-" * 70)

# Convert input and kernel to RNS
print("Converting input tensor to RNS...")
input_rns = tensor_to_rns(input_tensor, moduli)
print(f"Input RNS shape: {input_rns.shape}")

print("Converting kernel to RNS...")
kernel_rns = tensor_to_rns(kernel, moduli)
print(f"Kernel RNS shape: {kernel_rns.shape}")

# Perform convolution in RNS domain
print("\nPerforming convolution in RNS domain...")
conv_output_rns = conv2d_rns(input_rns, kernel_rns, moduli)
print(f"Convolution output RNS shape: {conv_output_rns.shape}")

# Convert RNS output back to real domain (unsigned for comparison)
conv_output_from_rns = tensor_from_rns(conv_output_rns, moduli, signed=False)
print(f"\nConvolution output (converted from RNS, unsigned):")
print(conv_output_from_rns)

# Apply ReLU in RNS domain
print("\nApplying ReLU in RNS domain...")
relu_output_rns = relu_rns(conv_output_rns, moduli)

# Convert ReLU output back to real domain
relu_output_from_rns = tensor_from_rns(relu_output_rns, moduli, signed=False)
print(f"ReLU output (converted from RNS):")
print(relu_output_from_rns)

# ============================================================
# Comparison of Results
# ============================================================

print("\n" + "-" * 70)
print("COMPARISON OF REAL DOMAIN vs RNS DOMAIN")
print("-" * 70)

print("\n1. CONVOLUTION OUTPUT COMPARISON:")
print("-" * 40)
print(f"{'Position':<12} {'Real (mod M)':<15} {'RNS':<15} {'Match'}")
print("-" * 50)

conv_match_count = 0
total_positions = conv_output_real.size

for i in range(conv_output_real.shape[0]):
    for j in range(conv_output_real.shape[1]):
        real_val = int(conv_output_real_mod[i, j])
        rns_val = int(conv_output_from_rns[i, j])
        match = "✓" if real_val == rns_val else "✗"
        if real_val == rns_val:
            conv_match_count += 1
        print(f"({i},{j}){'':<8} {real_val:<15} {rns_val:<15} {match}")

print(f"\nConvolution match rate: {conv_match_count}/{total_positions} ({100*conv_match_count/total_positions:.1f}%)")

print("\n2. ReLU OUTPUT COMPARISON:")
print("-" * 40)

# For ReLU comparison, we need to apply ReLU to the signed interpretation
# First get signed values from RNS
conv_output_signed = tensor_from_rns(conv_output_rns, moduli, signed=True)
relu_expected = np.maximum(0, conv_output_signed)

print(f"{'Position':<12} {'Real ReLU':<15} {'RNS ReLU':<15} {'Match'}")
print("-" * 50)

relu_match_count = 0

for i in range(relu_output_real.shape[0]):
    for j in range(relu_output_real.shape[1]):
        # Get signed value from RNS for ReLU comparison
        signed_val = int(conv_output_signed[i, j])
        expected_relu = max(0, signed_val)
        rns_relu_val = int(relu_output_from_rns[i, j])

        match = "✓" if expected_relu == rns_relu_val else "✗"
        if expected_relu == rns_relu_val:
            relu_match_count += 1
        print(f"({i},{j}){'':<8} {expected_relu:<15} {rns_relu_val:<15} {match}")

print(f"\nReLU match rate: {relu_match_count}/{total_positions} ({100*relu_match_count/total_positions:.1f}%)")

# ============================================================
# Detailed Analysis
# ============================================================

print("\n" + "-" * 70)
print("DETAILED ANALYSIS")
print("-" * 70)

print("\n1. Sample RNS Representation:")
print("-" * 40)
sample_i, sample_j = 0, 0
print(f"Position ({sample_i}, {sample_j}):")
print(f"  Real domain value: {conv_output_real[sample_i, sample_j]}")
print(f"  Real domain value mod {M}: {conv_output_real_mod[sample_i, sample_j]}")
print(f"  RNS representation: {list(conv_output_rns[sample_i, sample_j, :])}")
print(f"  RNS converted back: {conv_output_from_rns[sample_i, sample_j]}")

print("\n2. Overflow Analysis:")
print("-" * 40)
max_possible_value = 256 * 256 * 5 * 5 * 3  # max input * max weight * kernel size
print(f"Maximum possible convolution value: ±{max_possible_value}")
print(f"Dynamic range M: {M}")
print(f"Overflow expected: {'Yes' if max_possible_value > M else 'No'}")
print(f"\nNote: Since max possible value ({max_possible_value}) > M ({M}),")
print(f"      overflow occurs and results are computed modulo M.")
print(f"      This is why we compare real domain results mod M with RNS results.")

# ============================================================
# Summary
# ============================================================

print("\n" + "=" * 70)
print("PART 5 SUMMARY")
print("=" * 70)
print(f"""
CNN Configuration:
  - Input: 8×8×3
  - Kernel: 5×5×3
  - Output: 4×4×1 (after convolution)
  - Activation: ReLU

RNS Configuration:
  - Moduli: {moduli}
  - Dynamic Range M: {M}
  - Value Range: [-256, 255] for inputs/weights

Results:
  - Convolution outputs match (mod M): {conv_match_count}/{total_positions}
  - ReLU outputs match: {relu_match_count}/{total_positions}

Key Observations:
  1. RNS allows parallel computation of convolution across moduli
  2. All arithmetic operations (multiply, add) work independently per modulus
  3. Results are inherently computed modulo M, preventing overflow issues
  4. ReLU requires conversion back to determine sign (can be optimized)
""")
print("=" * 70)
```
